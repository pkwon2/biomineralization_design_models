```python
# In[1]: initialize PyRosetta and import movers
from pyrosetta import init, pose_from_pdb
from pyrosetta.rosetta.protocols.simple_moves import MutateResidue
from pyrosetta.rosetta.protocols.minimization_packing import PackRotamersMover

# mute most of the Rosetta output
init(extra_options="-mute all")
```

```python
# In[2]: load source & target poses
src_pdb = "/home/pkwon2/cage/superfold/rmsd_filtered/" \
          "all_helical_fiber_value1_0_ckpt_BFF_0_0_refined_13_combined_" \
          "packed_12_1_model_3_ptm_seed_0_unrelaxed.pdb"

tgt_pdb = "/home/pkwon2/cage/1_diffusion/outputs/" \
          "all_helical_fiber_value1_0_ckpt_BFF_0_0_refined_13.pdb"

src_pose = pose_from_pdb(src_pdb)
tgt_pose = pose_from_pdb(tgt_pdb)

print(f"Loaded source ({len(src_pose.sequence())} aa) and "
      f"target ({len(tgt_pose.sequence())} aa)")
```

```python
# In[3]: mutate only Glu (E), Asp (D), Ser (S)

# map one‐letter → three‐letter codes
three = {'D':'ASP', 'E':'GLU', 'S':'SER'}

n = min(len(src_pose.sequence()), len(tgt_pose.sequence()))
for i in range(1, n+1):
    aa1 = src_pose.residue(i).name1()   # one‐letter at pos i
    if aa1 in three:
        res3 = three[aa1]
        mover = MutateResidue(i, res3)
        mover.apply(tgt_pose)
        print(f"  mutated pos {i} → {res3}")
```

```python
# In[4]: optional repack around the mutations to relieve clashes
pack = PackRotamersMover()
pack.apply(tgt_pose)
print("Repacked side‐chains")
```

```python
# In[5]: write out your threaded model
out_pdb = "threaded_EDS_only.pdb"
tgt_pose.dump_pdb(out_pdb)
print(f"Saved threaded model to {out_pdb}")
```
