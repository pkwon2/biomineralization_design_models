```python
import os
from Bio.PDB import PDBParser

def main():
    input_dir = os.path.abspath("./mpnn/input")
    os.makedirs("mpnn", exist_ok=True)
    tasks_file = os.path.join("mpnn", "tasks")

    parser = PDBParser(QUIET=True)
    # Residue names to exclude from grouping and redesign
    exclude_resnames = {"GLU", "ASP", "LYS", "ARG"}

    with open(tasks_file, "w") as out:
        for fname in os.listdir(input_dir):
            if not fname.endswith(".pdb"):
                continue
            pdb_path = os.path.join(input_dir, fname)

            # Parse PDB and compute chain A length
            structure = parser.get_structure(fname, pdb_path)
            model = structure[0]
            chain_lengths = {}
            if "A" in model:
                resnums = [r.id[1] for r in model["A"].get_residues() if r.id[0] == " "]
                if resnums:
                    chain_lengths["A"] = max(resnums)

            # Divide chain A into 4 equal segments and group corresponding residues
            all_groups = []
            if "A" in chain_lengths:
                length_A = chain_lengths["A"]
                repeat_len = length_A // 4
                for i in range(repeat_len):
                    labels = []
                    for j in range(4):
                        resnum = i + 1 + j * repeat_len
                        if resnum <= length_A:
                            labels.append(f"A{resnum}")
                    if len(labels) == 4:
                        all_groups.append(",".join(labels))

            # Filter symmetry groups to exclude specified residues
            grouped_labels = []
            for grp in all_groups:
                for label in grp.split(","):
                    chain = label[0]
                    resnum = int(label[1:])
                    try:
                        res = model[chain][resnum]
                    except KeyError:
                        continue
                    if res.get_resname() in exclude_resnames:
                        continue
                    grouped_labels.append(label)

            # Add all other non-excluded residues
            all_labels = []
            for chain in model:
                for res in chain.get_residues():
                    if res.id[0] != " ":  # skip HETATM/insertion codes
                        continue
                    if res.get_resname() in exclude_resnames:
                        continue
                    label = f"{chain.id}{res.id[1]}"
                    if label not in grouped_labels:
                        all_labels.append(label)

            redesigned_residues = " ".join(grouped_labels + all_labels)
            symmetry_residues = "|".join(all_groups)
            symmetry_weights = "|".join(
                ",".join(["1"] * len(grp.split(","))) for grp in all_groups
            )

            # Construct and write command
            cmd = (
                "/software/containers/mlfold.sif /databases/mpnn/fused_mpnn/run.py "
                "--temperature 0.3 "
                f"--pdb_path {pdb_path} "
                "--out_folder outputs "
                "--batch_size 5 "
                "--number_of_batches 5 "
                "--bias_AA 'M:-10.0' "
                "--pack_side_chains 1 "
                "--repack_everything 1 "
                f"--redesigned_residues '{redesigned_residues}' "
                f"--symmetry_residues '{symmetry_residues}' "
                f"--symmetry_weights '{symmetry_weights}'"
            )
            out.write(cmd + "\n")

    print(f"Tasks file written to {tasks_file}")

if __name__ == "__main__":
    main()
```

```python
import os
from Bio import PDB

def find_non_charged_residues(file_path):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure('PDB_structure', file_path)
    residue_positions = []
    
    # Define the list of charged residues
    charged_residues = ['ASP', 'GLU', 'SER']

    for model in structure:
        for chain in model:
            # Get all residues in the chain as a list
            residues_list = list(chain.get_residues())
            # Exclude the last 88 residues from processing
            if len(residues_list) > 1:
                residues_to_process = residues_list[:]
            else:
                residues_to_process = []  # Skip if the chain is too short

            for residue in residues_to_process:
                resname = residue.get_resname()
                residue_number = residue.id[1]
                # Exclude residues in positions 1-27 and 35-81
                # if 1 <= residue_number <= 27 or 35 <= residue_number <= 81:
                #     continue
                # Add if the residue is not charged
                if resname not in charged_residues:
                    residue_positions.append(f"{chain.id}{residue_number}")
                    
    return ' '.join(residue_positions)

def main():
    pdb_dir = './mpnn/input'
    
    # Convert relative path to absolute path
    pdb_dir = os.path.abspath(pdb_dir)

    if not os.path.exists(pdb_dir):
        print(f"Directory {pdb_dir} does not exist.")
        return

    # Base directory for output tasks
    base_directory = "mpnn"
    os.makedirs(base_directory, exist_ok=True)

    # Define the base command with updated placeholder {residues}
    base_command = "/software/containers/mlfold.sif /databases/mpnn/fused_mpnn/run.py " \
                   "--temperature 0.3 " \
                   "--redesigned_residues '{residues}' " \
                   "--homo_oligomer 1 " \
                   "--pdb_path {file_path} " \
                   "--out_folder outputs " \
                   "--batch_size 4 " \
                   "--number_of_batches 4 " \
                   "--bias_AA 'M:-10.0' " \
                   "--pack_side_chains 1 " \
                   "--repack_everything 1"

    tasks_file = os.path.join(base_directory, "tasks")

    with open(tasks_file, "w") as file:
        for pdb_file in os.listdir(pdb_dir):
            if pdb_file.endswith('.pdb'):
                file_path = os.path.join(pdb_dir, pdb_file)
                residues = find_non_charged_residues(file_path)
                command = base_command.format(file_path=file_path, residues=residues)
                file.write(f"{command}\n")

    print(f"Tasks file '{tasks_file}' has been created.")

if __name__ == "__main__":
    main()
```

```python
import os
from Bio.PDB import PDBParser, PPBuilder
from Bio.SeqIO import write
from Bio.SeqRecord import SeqRecord
from Bio.Seq import Seq

# Define input and output directories
pdb_dir = "/home/pkwon2/cage/mpnn/outputs/packed"
fasta_dir = "/home/pkwon2/cage/mpnn/outputs/fasta"

# Ensure output directory exists
os.makedirs(fasta_dir, exist_ok=True)

# Initialize PDB parser and peptide builder
parser = PDBParser(QUIET=True)
ppb = PPBuilder()

# Process each PDB file
for pdb_file in os.listdir(pdb_dir):
    if pdb_file.endswith(".pdb"):
        pdb_path = os.path.join(pdb_dir, pdb_file)
        structure_id = os.path.splitext(pdb_file)[0]
        structure = parser.get_structure(structure_id, pdb_path)

        chain_sequences = []
        for model in structure:
            for chain in model:
                peptides = ppb.build_peptides(chain)
                chain_seq = "".join(str(pp.get_sequence()) for pp in peptides)
                if chain_seq:
                    chain_sequences.append(chain_seq.strip())
            break  # Only process the first model

        if chain_sequences:
            combined_sequence = " ".join(chain_sequences).strip()
            fasta_path = os.path.join(fasta_dir, f"{structure_id}.fasta")
            with open(fasta_path, "w") as f:
                f.write(f">{structure_id}\n{combined_sequence}\n")
```

```python
#!/usr/bin/env python3
import os
import glob
import shutil
from Bio.PDB import MMCIFParser, PDBIO

def convert_cif_to_pdb(cif_file, pdb_file):
    """
    Convert a CIF file to a PDB file using BioPython.
    """
    parser = MMCIFParser(QUIET=True)
    structure = parser.get_structure("structure", cif_file)
    io = PDBIO()
    io.set_structure(structure)
    io.save(pdb_file)

def main():
    # Define directories
    outputs_dir = "./alphafold3/outputs"
    pdb_dir = os.path.join(outputs_dir, "pdb")
    
    # Create pdb directory if it doesn't exist
    os.makedirs(pdb_dir, exist_ok=True)

    # Find all CIF files in subdirectories of outputs
    cif_pattern = os.path.join(outputs_dir, "*", "*.cif")
    cif_files = glob.glob(cif_pattern)
    
    # Find all summary_confidences.json files in subdirectories of outputs
    json_pattern = os.path.join(outputs_dir, "*", "*summary_confidences.json")
    json_files = glob.glob(json_pattern)

    # Process CIF files: convert to pdb and save in pdb_dir
    for cif in cif_files:
        base_name = os.path.basename(cif)
        pdb_filename = os.path.splitext(base_name)[0] + ".pdb"
        pdb_path = os.path.join(pdb_dir, pdb_filename)
        try:
            convert_cif_to_pdb(cif, pdb_path)
            print(f"Converted {cif} to {pdb_path}")
        except Exception as e:
            print(f"Failed to convert {cif}: {e}")

    # Process JSON files: copy them to pdb_dir instead of moving
    for json in json_files:
        dest = os.path.join(pdb_dir, os.path.basename(json))
        try:
            shutil.copy2(json, dest)
            print(f"Copied {json} to {dest}")
        except Exception as e:
            print(f"Failed to copy {json}: {e}")

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
import os
import glob
from Bio.PDB import PDBParser, PDBIO, Select

# --- parameters ---
input_dir  = "/home/pkwon2/cage/alphafold3/filtered"
output_dir = "/home/pkwon2/cage/alphafold3/order"
os.makedirs(output_dir, exist_ok=True)

parser = PDBParser(QUIET=True)
io     = PDBIO()

class QuarterSelect(Select):
    """
    Only accept atoms whose (chain, resid) is in the keep_set.
    """
    def __init__(self, keep_set):
        self.keep = keep_set

    def accept_atom(self, atom):
        chain_id = atom.get_parent().get_parent().id
        resid    = atom.get_parent().get_id()
        # resid is a tuple: (hetfield, resseq, icode)
        # we only included those with hetfield==' '
        return (chain_id, resid) in self.keep

for pdb_path in glob.glob(os.path.join(input_dir, "*.pdb")):
    name      = os.path.basename(pdb_path)
    structure = parser.get_structure(name, pdb_path)
    model     = structure[0]

    # 1) collect all standard residues (hetfield==' ')
    all_res = []
    for chain in model:
        for res in chain:
            hetfield, resseq, icode = res.get_id()
            if hetfield == " ":
                all_res.append((chain.id, res.get_id()))

    total = len(all_res)
    if total == 0:
        print(f"WARNING: no standard residues found in {name}, skipping.")
        continue

    # 2) compute cutoff
    cutoff = total // 4
    keep_set = set(all_res[:cutoff])

    # 3) write trimmed PDB
    io.set_structure(structure)
    out_path = os.path.join(output_dir, name)
    io.save(out_path, select=QuarterSelect(keep_set))

    print(f"{name}: kept {cutoff}/{total} residues → {out_path}")
```

```python
#!/usr/bin/env python3
import os
import glob
import shutil

# --- CONFIGURE ---
CIF_DIR       = "/home/pkwon2/cage/alphafold3/filtered"
FASTA_DIR     = "/home/pkwon2/cage/mpnn/outputs/fasta"
UNPAIRED_DIR  = os.path.join(FASTA_DIR, "unpaired")
# -----------------

def split_into_four(seq: str):
    """
    Split `seq` into 4 as-equal-as-possible segments,
    distributing any extra residues into the earliest segments.
    """
    n = len(seq)
    base, rem = divmod(n, 4)
    segments = []
    start = 0
    for i in range(4):
        seg_len = base + (1 if i < rem else 0)
        segments.append(seq[start:start+seg_len])
        start += seg_len
    return segments

def process_fasta(fasta_path: str):
    # Read & concatenate all sequence lines
    with open(fasta_path) as f:
        lines = f.read().splitlines()
    header = lines[0]
    seq = "".join(lines[1:]).replace(" ", "")

    # Split into four segments and re-join with 3 spaces
    segments = split_into_four(seq)
    new_seq = " ".join(segments)

    # Overwrite FASTA
    with open(fasta_path, "w") as f:
        f.write(f"{header}\n{new_seq}\n")

def main():
    # 1) Build set of prefixes from CIF files
    cif_pattern = os.path.join(CIF_DIR, "*_model.cif")
    cif_prefixes = {
        os.path.basename(p).rsplit("_model.cif", 1)[0]
        for p in glob.glob(cif_pattern)
    }

    # Ensure unpaired directory exists
    os.makedirs(UNPAIRED_DIR, exist_ok=True)

    # 2) Iterate through all FASTAs
    fasta_pattern = os.path.join(FASTA_DIR, "*.fasta")
    for fasta_path in glob.glob(fasta_pattern):
        fname = os.path.basename(fasta_path)
        prefix = fname.rsplit(".fasta", 1)[0]

        if prefix in cif_prefixes:
            # paired → process in place
            process_fasta(fasta_path)
            print(f"✔ Paired & updated: {fname}")
        else:
            # unpaired → move out
            dest = os.path.join(UNPAIRED_DIR, fname)
            shutil.move(fasta_path, dest)
            print(f"→ Moved unpaired: {fname} → unpaired/")

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
import os
import json
import glob

# === Configuration ===
filtered_dir = '/home/pkwon2/cage/alphafold3/filtered'
TOP_N = 2   # keep the top N scoring entries per group

# === Build groups by prefix before last "packed" ===
groups = {}
pattern = os.path.join(filtered_dir, '*_summary_confidences.json')
for js_path in glob.glob(pattern):
    fname = os.path.basename(js_path)
    base  = fname[:-len('_summary_confidences.json')]   # e.g. "XYZ_packed_4_1"
    idx   = base.lower().rfind('packed')
    if idx == -1:
        continue
    prefix = base[:idx].rstrip('_')
    groups.setdefault(prefix, []).append(base)

# === For each group, keep top TOP_N by "ranking_score", delete the rest ===
for prefix, members in groups.items():
    scores = []
    for m in members:
        js = os.path.join(filtered_dir, f"{m}_summary_confidences.json")
        try:
            conf = json.load(open(js))
            raw = conf.get('ranking_score')
            score = float(raw) if raw is not None else 0.0
        except Exception:
            score = 0.0
        scores.append((m, score))

    # sort descending, pick top TOP_N
    scores_sorted = sorted(scores, key=lambda x: x[1], reverse=True)
    keep = {m for m, _ in scores_sorted[:TOP_N]}

    # delete subgroup files not in keep
    for m, _ in scores:
        if m in keep:
            continue
        for fname in (
            f"{m}.pdb",
            f"{m}_model.cif",
            f"{m}_summary_confidences.json",
        ):
            fp = os.path.join(filtered_dir, fname)
            if os.path.exists(fp):
                os.remove(fp)
                print(f"Deleted {fp}")

print("Pruning complete.")
```

```python
#!/usr/bin/env python3
import os
from glob import glob

# === Paths ===
in_dir  = '/home/pkwon2/cage/1_diffusion/input_post_af3'
out_dir = '/home/pkwon2/cage/1_diffusion/input_post_af3_trimmed'
os.makedirs(out_dir, exist_ok=True)

for pdb_fp in glob(os.path.join(in_dir, '*.pdb')):
    # Read all lines
    lines = open(pdb_fp).read().splitlines()

    # Group ATOM/HETATM lines by chain in order of appearance
    chain_order = []
    chain_lines = {}
    for l in lines:
        if l.startswith(('ATOM','HETATM')):
            chain = l[21]
            if chain not in chain_order:
                chain_order.append(chain)
                chain_lines[chain] = []
            chain_lines[chain].append(l)

    # Process each chain: drop first 35 & last 30 residues, renumber from 1
    out_lines = []
    for chain in chain_order:
        seqs = []
        for l in chain_lines[chain]:
            res = int(l[22:26])
            if res not in seqs:
                seqs.append(res)

        # Determine which residues to keep
        if len(seqs) > 35 + 30:
            keep = seqs[35:len(seqs)-30]
        else:
            keep = []

        # Map old→new residue numbers
        seq_map = {old: new for new, old in enumerate(keep, start=1)}

        # Write kept lines with updated residue numbers
        for l in chain_lines[chain]:
            old = int(l[22:26])
            if old in seq_map:
                new = seq_map[old]
                out_lines.append(f"{l[:22]}{new:>4}{l[26:]}")

    # Write trimmed PDB
    out_fp = os.path.join(out_dir, os.path.basename(pdb_fp))
    with open(out_fp, 'w') as fw:
        fw.write('\n'.join(out_lines) + '\n')
```

```python
# Specify parameters in a notebook cell
input_dir = "/home/pkwon2/cage/1_diffusion/input_post_af3_trimmed"
output_dir = "/home/pkwon2/cage/1_diffusion/input_post_af3_multichain"
copies = 5  # number of additional chains (e.g., 3 → B, C, D)

import os
import glob
from Bio.PDB import PDBParser, PDBIO, Superimposer

# Ensure output directory exists
os.makedirs(output_dir, exist_ok=True)

# Initialize parser, I/O handler, and chain ID list
parser = PDBParser(QUIET=True)
io = PDBIO()
chain_ids = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'

for pdb_path in glob.glob(os.path.join(input_dir, '*.pdb')):
    name = os.path.basename(pdb_path)
    structure = parser.get_structure('orig', pdb_path)
    model = structure[0]
    chain_A = model['A']

    # Collect only standard residues
    residues = [r for r in chain_A if r.get_id()[0] == ' ']
    if len(residues) < 10:
        print(f"Skipping {name}: fewer than 10 residues in chain A")
        continue

    first10 = residues[:10]
    last10 = residues[-10:]

    # Extract CA atoms for alignment
    fixed_atoms = [r['CA'] for r in last10]
    mobile_atoms = [r['CA'] for r in first10]

    # Compute superposition
    sup = Superimposer()
    sup.set_atoms(fixed_atoms, mobile_atoms)
    rot, trans = sup.rotran

    print(f"\n=== {name} ===")
    print("Rotation matrix:\n", rot)
    print("Translation vector:\n", trans)

    # Generate specified number of copies
    for i in range(1, copies + 1):
        copy_struct = parser.get_structure(f'mob{i}', pdb_path)
        copy_model = copy_struct[0]
        copy_chain = copy_model['A']
        # Apply transform i times
        for _ in range(i):
            sup.apply(list(copy_chain.get_atoms()))
        # Assign new chain ID
        if i < len(chain_ids):
            copy_chain.id = chain_ids[i]
        else:
            raise ValueError(f"Exceeded chain ID limit at copy {i}")
        # Append to original model
        model.add(copy_chain)

    # Save combined structure
    out_file = os.path.join(output_dir, name)
    io.set_structure(structure)
    io.save(out_file)
    print(f"Wrote multi-chain PDB to {out_file}")
```

```python
#!/usr/bin/env python3
"""
Remove the last 10 residues from every chain in each PDB file and overwrite the original.

This script parses each PDB in `pdb_dir`, deletes the final ten standard residues (by sequence order) from every chain,
then writes the modified structure back to the same path.
"""
import os
import glob
from Bio.PDB import PDBParser, PDBIO

# === CONFIGURATION ===
# Directory containing multi-chain PDBs (e.g., output of multichain_transform)
pdb_dir = "/home/pkwon2/cage/1_diffusion/input_post_af3_multichain"

# Ensure directory exists
os.makedirs(pdb_dir, exist_ok=True)

# Initialize parser and I/O handler
parser = PDBParser(QUIET=True)
io = PDBIO()

# Process each PDB file
for pdb_path in glob.glob(os.path.join(pdb_dir, "*.pdb")):
    filename = os.path.basename(pdb_path)
    structure = parser.get_structure('struct', pdb_path)
    model = structure[0]

    # Iterate through each chain in the model
    for chain in list(model):
        # Select only standard residues
        residues = [res for res in chain if res.get_id()[0] == ' ']
        if len(residues) <= 10:
            print(f"Skipping chain {chain.id} in {filename}: <=10 residues.")
            continue

        # Identify and remove the last 10 residues
        to_remove = residues[-10:]
        for res in to_remove:
            chain.detach_child(res.get_id())

    # Overwrite the original PDB with the trimmed structure
    io.set_structure(structure)
    io.save(pdb_path)

print("Done trimming last 10 residues from all chains in each PDB.")
```

```python
#!/usr/bin/env python3
"""
Filter PDBs in-place by keeping only the first 60 residues of chain B
and the last 60 residues of chain A, deleting other residues in those chains.
All other chains remain unmodified.
"""
import glob
import os

# Directory containing combined PDBs
pdb_dir = "/home/pkwon2/cage/1_diffusion/for_ligand_mpnn"

# Helper to parse residue number from PDB line
def get_chain_resnum(line):
    chain = line[21]
    try:
        resnum = int(line[22:26].strip())
    except ValueError:
        return chain, None
    return chain, resnum

# Process each PDB
for pdb_path in glob.glob(os.path.join(pdb_dir, "*_combined.pdb")):
    # Read all lines and collect residue numbers for chains A and B
    residues = {'C': set(), 'B': set()}
    lines = []
    with open(pdb_path, 'r') as f:
        for line in f:
            lines.append(line)
            if line.startswith(('ATOM  ', 'HETATM')):
                chain, resnum = get_chain_resnum(line)
                if chain in residues and resnum is not None:
                    residues[chain].add(resnum)

    # Determine residues to keep
    sorted_C = sorted(residues['C'])
    sorted_B = sorted(residues['B'])
    keep_C = set(sorted_C[-75:]) if len(sorted_C) >= 75 else set(sorted_C)
    keep_B = set(sorted_B[:102]) if len(sorted_B) >= 102 else set(sorted_B)

    # Write filtered PDB back to same file
    with open(pdb_path, 'w') as fout:
        for line in lines:
            if line.startswith(('ATOM  ', 'HETATM')):
                chain, resnum = get_chain_resnum(line)
                if chain == 'C':
                    if resnum in keep_C:
                        fout.write(line)
                elif chain == 'B':
                    if resnum in keep_B:
                        fout.write(line)
                else:
                    fout.write(line)
            else:
                fout.write(line)

    print(f"Overwritten {pdb_path}")
```

```python
# Filter Nearest Hetero-Residues Jupyter Notebook

# Cell 1: Imports and Setup
import os
import glob
import math

# Cell 2: Helper Functions

def read_pdb_coords(lines):
    """Parse CA coordinates from ATOM records and group HETATM by residue."""
    protein_coords = []
    het_groups = {}

    for line in lines:
        record = line[:6]
        # Only consider C-alpha atoms for protein coordinates
        if record == "ATOM  " and line[12:16].strip() == "CA":
            x, y, z = map(float, (line[30:38], line[38:46], line[46:54]))
            protein_coords.append((x, y, z))

        elif record == "HETATM":
            # Group hetero atoms by residue identifier
            chain   = line[21]
            resseq  = int(line[22:26])
            icode   = line[26]
            resname = line[17:20].strip()
            key     = (chain, resseq, icode, resname)

            coord = tuple(map(float, (line[30:38], line[38:46], line[46:54])))
            grp = het_groups.setdefault(key, {"coords": [], "lines": []})
            grp["coords"].append(coord)
            grp["lines"].append(line)

    return protein_coords, het_groups


def compute_min_dist(protein_coords, het_coords):
    """Return the minimum Euclidean distance between het_coords and protein_coords."""
    min_d2 = float('inf')
    for hx, hy, hz in het_coords:
        for px, py, pz in protein_coords:
            dx, dy, dz = hx - px, hy - py, hz - pz
            d2 = dx*dx + dy*dy + dz*dz
            if d2 < min_d2:
                min_d2 = d2
    return math.sqrt(min_d2) if min_d2 < float('inf') else float('inf')


def filter_pdb(pdb_path, output_dir, num_keep, max_dist):
    """Filter a single PDB, keeping only the nearest het residues to CA atoms within max_dist."""
    with open(pdb_path) as f:
        lines = f.readlines()

    prot_coords, het_groups = read_pdb_coords(lines)

    # Compute the distance of each hetero-residue to any CA atom
    dists = []
    for key, grp in het_groups.items():
        d = compute_min_dist(prot_coords, grp['coords'])
        dists.append((d, key))

    # Filter by max_dist and select nearest num_keep
    eligible = [item for item in dists if item[0] <= max_dist]
    eligible.sort(key=lambda x: x[0])
    keep_keys = {key for _, key in eligible[:num_keep]}

    out_path = os.path.join(output_dir, os.path.basename(pdb_path))
    with open(out_path, 'w') as out:
        for line in lines:
            if line[:6] == 'HETATM':
                chain   = line[21]
                resseq  = int(line[22:26])
                icode   = line[26]
                resname = line[17:20].strip()
                key     = (chain, resseq, icode, resname)
                if key in keep_keys:
                    out.write(line)
            else:
                out.write(line)

    kept = len(keep_keys)
    print(f"Processed {os.path.basename(pdb_path)}: kept {kept} hetero-residues within {max_dist} Å")

# Cell 3: Parameters and Execution
input_dir  = "/home/pkwon2/cage/1_diffusion/for_ligand_mpnn"
output_dir = os.path.join(input_dir, "filtered_het10")
num_keep   = 1000
max_dist   = 9.0  # maximum distance threshold in Angstroms

os.makedirs(output_dir, exist_ok=True)

pdb_files = glob.glob(os.path.join(input_dir, "*.pdb"))
for pdb in pdb_files:
    filter_pdb(pdb, output_dir, num_keep, max_dist)
```

```python
#!/usr/bin/env python3
import os
import glob
from Bio.PDB import PDBParser, PDBIO, Select

class ChainASelect(Select):
    """Select only chain A."""
    def accept_chain(self, chain):
        return chain.id == 'A'

def main():
    order_dir = "/home/pkwon2/cage/alphafold3/filtered"
    cowboy_dir = "/home/pkwon2/cage/cowboy"
    os.makedirs(cowboy_dir, exist_ok=True)

    # Get all PDB files in the order directory
    pdb_files = glob.glob(os.path.join(order_dir, "*.pdb"))
    parser = PDBParser(QUIET=True)
    
    for pdb_file in pdb_files:
        structure = parser.get_structure(os.path.basename(pdb_file), pdb_file)
        io = PDBIO()
        io.set_structure(structure)
        output_path = os.path.join(cowboy_dir, os.path.basename(pdb_file))
        try:
            io.save(output_path, select=ChainASelect())
            print(f"Copied chain A from {pdb_file} to {output_path}")
        except Exception as e:
            print(f"Failed to copy chain A from {pdb_file}: {e}")

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
import os
import glob
import csv
import sys

def csv_to_fasta(csv_dir, output_filename="designs_pET29b+_cloned.fasta"):
    # Find CSV files in the directory
    pattern = os.path.join(csv_dir, "*.csv")
    csv_files = glob.glob(pattern)
    if not csv_files:
        print(f"No CSV files found in {csv_dir}", file=sys.stderr)
        sys.exit(1)

    # Use the first CSV found
    csv_path = csv_files[0]
    fasta_path = os.path.join(csv_dir, output_filename)

    with open(csv_path, newline='') as csvfile, open(fasta_path, 'w') as fastafile:
        reader = csv.DictReader(csvfile)
        if 'design_name' not in reader.fieldnames or 'ORF_from_LM0627' not in reader.fieldnames:
            print("CSV must contain 'design_name' and 'ORF_from_LM0627' columns", file=sys.stderr)
            sys.exit(1)

        for row in reader:
            name = row['design_name'].strip()
            seq = row['ORF_from_LM0627'].strip()

            # Remove leading 'atg' (case-insensitive)
            if seq.lower().startswith('atg'):
                seq = seq[3:]

            # Write to FASTA
            fastafile.write(f">{name}\n")
            fastafile.write(f"{seq}\n")

    print(f"Wrote FASTA to {fasta_path}")

if __name__ == "__main__":
    # Default directory; adjust if needed
    csv_directory = "/home/pkwon2/cage/cowboy"
    csv_to_fasta(csv_directory)
```

```python
#!/usr/bin/env python3
import os
import glob
import numpy as np

def get_ca_coords(pdb_path):
    """
    Parse the PDB file and return an (N,3) array of C-alpha coordinates
    for all chains except A, B, and C.
    """
    coords = []
    with open(pdb_path) as f:
        for line in f:
            if line.startswith("ATOM") and line[12:16].strip() == "CA":
                chain_id = line[21].strip()
                if chain_id in ("A", "B", "C"):
                    continue
                x = float(line[30:38])
                y = float(line[38:46])
                z = float(line[46:54])
                coords.append((x, y, z))
    return np.array(coords)

def compute_flatness(ca_coords):
    """
    Fit a plane to the points via PCA: compute covariance matrix,
    find eigenvector with smallest eigenvalue (plane normal),
    then compute RMS distance of points to that plane.
    """
    centroid = ca_coords.mean(axis=0)
    pts_centered = ca_coords - centroid

    cov = np.cov(pts_centered, rowvar=False)
    eigvals, eigvecs = np.linalg.eigh(cov)
    normal = eigvecs[:, np.argmin(eigvals)]

    dists = np.dot(pts_centered, normal)
    rms_dev = np.sqrt(np.mean(dists**2))
    return rms_dev

def main():
    folder = "/home/pkwon2/cage/alphafold3/filtered"
    pattern = os.path.join(folder, "*_model.pdb")
    results = []

    for pdb_file in glob.glob(pattern):
        ca = get_ca_coords(pdb_file)
        if ca.shape[0] < 3:
            print(f"Skipping {os.path.basename(pdb_file)} (fewer than 3 filtered Cα atoms)")
            continue
        flatness = compute_flatness(ca)
        results.append((os.path.basename(pdb_file), flatness))

    # sort ascending: flattest first
    results.sort(key=lambda x: x[1])

    print("PDB file\tFlatness (RMS Å)")
    print("--------\t---------------")
    for name, rms in results:
        print(f"{name}\t{rms:.3f}")

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
import os
import csv
import shutil
from glob import glob
import pymol2

# === Paths ===
af_pdb_dir    = '/home/pkwon2/cage/alphafold3/outputs/pdb'
ref_pdb       = '/home/pkwon2/cage/1_diffusion/input/tile_b_82_packed_2_1.pdb'
filtered_dir  = '/home/pkwon2/cage/alphafold3/filtered'
out_csv       = os.path.join(filtered_dir, 'tile_b_82_rmsd.csv')

os.makedirs(filtered_dir, exist_ok=True)

with open(out_csv, 'w', newline='') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(['model_filename', 'rmsd_CA'])

    with pymol2.PyMOL() as pymol:
        cmd = pymol.cmd
        cmd.load(ref_pdb, 'ref')

        for model_fp in glob(os.path.join(af_pdb_dir, '*tile_b_82*.pdb')):
            model_name = os.path.basename(model_fp)
            base       = os.path.splitext(model_name)[0]

            # strip off everything after the last "_model"
            prefix = base.rsplit('_model', 1)[0]
            json_fp = os.path.join(af_pdb_dir, prefix + '_summary_confidences.json')

            # load model
            cmd.load(model_fp, 'm1')

            # run cealign on CA atoms
            ce_res = cmd.cealign('m1 and name CA', 'ref and name CA')

            # extract RMSD, depending on return type
            if isinstance(ce_res, (list, tuple)) and len(ce_res) >= 1:
                rmsd = float(ce_res[0])
            elif isinstance(ce_res, dict) and 'RMSD' in ce_res:
                rmsd = float(ce_res['RMSD'])
            else:
                # if neither format matches, set RMSD = 0.0 (or handle as you like)
                rmsd = 0.0

            # delete the model object before writing out
            cmd.delete('m1')

            writer.writerow([model_name, f"{rmsd:.3f}"])

            if rmsd < 0.9:
                shutil.copy2(model_fp, filtered_dir)
                if os.path.isfile(json_fp):
                    shutil.copy2(json_fp, filtered_dir)
                else:
                    print(f"Warning: JSON not found for {model_name} → looked for {json_fp}")

        cmd.delete('ref')

print(f"Done. RMSD values written to {out_csv}")
```

```python
#!/usr/bin/env python3
import os
import string

# === CONFIGURATION ===
input_dir  = "/home/pkwon2/cage/align/original"
output_dir = "/home/pkwon2/cage/align/renamed"
chain_ids  = list(string.ascii_uppercase)  # ['A', 'B', 'C', …, 'Z']
# =====================

os.makedirs(output_dir, exist_ok=True)

for fname in os.listdir(input_dir):
    if not fname.lower().endswith(".pdb"):
        continue

    in_path  = os.path.join(input_dir,  fname)
    out_path = os.path.join(output_dir, fname)

    with open(in_path, "r") as infile, open(out_path, "w") as outfile:
        letter_idx = -1
        current_chain = None

        for line in infile:
            # When we hit a TER record, advance to the next chain letter
            if line.startswith("TER"):
                letter_idx += 1
                if letter_idx >= len(chain_ids):
                    raise RuntimeError(
                        f"Too many MODEL sections in {fname}; no chain ID available beyond '{chain_ids[-1]}'"
                    )
                current_chain = chain_ids[letter_idx]
                # Do not write the MODEL line itself
                continue

            # Skip any ENDMDL, END, or further MODEL lines
            if line.startswith("ENDMDL") or line.startswith("MODEL"):
                continue

            # Only write lines once we have a chain letter assigned
            if current_chain is None:
                continue

            # Replace chain ID at column 22 (index 21) if the line is long enough
            if len(line) >= 22:
                new_line = line[:21] + current_chain + line[22:]
                outfile.write(new_line)

        # end for each line

    # end with infile/outfile
# end for each file

print(f"Processed all PDBs from '{input_dir}' → '{output_dir}'.")
```
