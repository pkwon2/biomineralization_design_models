```python
import os
import json
import pandas as pd
from pathlib import Path

# Define the directory containing the .json files
input_directory = Path('./dhr31_superfold/outputs')

# Initialize a list to store the data
data = []

# Loop through all .json files in the directory
for json_file in input_directory.glob('*.json'):
    # Read the JSON file
    with open(json_file, 'r') as file:
        content = json.load(file)
        
        # Extract mean_plddt and rmsd_to_input
        mean_plddt = content.get('mean_plddt', None)
        rmsd_to_input = content.get('rmsd_to_input', None)
        
        # Derive the corresponding .pdb file name
        pdb_file = str(json_file).replace('prediction_results.json', 'unrelaxed.pdb')
        
        # Get the base name of the .json file (without directory path)
        json_base_name = json_file.name
        
        # Append the data to the list
        data.append({
            'json_file': json_base_name,
            'mean_plddt': mean_plddt,
            'rmsd_to_input': rmsd_to_input,
            'pdb_file': pdb_file
        })

# Create a DataFrame from the collected data
df = pd.DataFrame(data)

# Define a function to extract the prefix before the last instance of 'packed' in the file name
def extract_prefix(file_name):
    parts = file_name.split('packed')
    if len(parts) > 1:
        return 'packed'.join(parts[:-1])
    return file_name

# Apply the function to create a new column for grouping
df['prefix'] = df['json_file'].apply(extract_prefix)

# Sort the DataFrame by mean_plddt in descending order
df_sorted = df.sort_values(by='mean_plddt', ascending=False)

# Group by the extracted prefix and keep the sorted order within each group
df_grouped_sorted = df_sorted.groupby('prefix', group_keys=False).apply(lambda x: x)

# Save the grouped DataFrame as a CSV file in the input directory
output_csv_path = input_directory / 'superfold_data_sorted.csv'
df_grouped_sorted.to_csv(output_csv_path, index=False)

# Display the grouped and sorted DataFrame
print(df_grouped_sorted)
```

```python
import os
import pandas as pd
from pathlib import Path
import shutil

# Define the paths
csv_path = Path('./dhr31_superfold/outputs/superfold_data_sorted.csv')
destination_directory = Path('./rigid_body_transformations/input')
destination_directory.mkdir(parents=True, exist_ok=True)

# Read the CSV file
df = pd.read_csv(csv_path)

# Function to copy files based on filtering criteria
def copy_files(df, best_per_prefix=False):
    if best_per_prefix:
        # Group by prefix and take the top 3 entries by mean_plddt
        df = df.groupby('prefix').apply(lambda x: x.nlargest(1, 'mean_plddt')).reset_index(drop=True)
    
    # Filter based on rmsd_to_input < 2 and mean_plddt > 0.8
    df_filtered = df[(df['rmsd_to_input'] < 4) & (df['mean_plddt'] > 0.8)]

    for _, row in df_filtered.iterrows():
        pdb_file = Path(row['pdb_file'])
        if pdb_file.exists():
            shutil.copy(pdb_file, destination_directory / pdb_file.name)
        else:
            print(f"File {pdb_file} does not exist.")

# Option to choose the filtering mode
choose_best_per_prefix = True  # Set to False to copy all files meeting the criteria
copy_files(df, best_per_prefix=choose_best_per_prefix)
```

```python
import os
import numpy as np
from scipy.spatial.transform import Rotation as R

# Function to parse PDB file and extract coordinates for specified residues
def parse_pdb(pdb_file, residue_ranges):
    coords = []
    with open(pdb_file, 'r') as file:
        for line in file:
            if line.startswith("ATOM") and line[12:16].strip() == "CA":
                chain = line[21]
                res_id = int(line[22:26].strip())
                for res_range in residue_ranges:
                    if chain == res_range[0] and res_range[1][0] <= res_id <= res_range[1][1]:
                        x = float(line[30:38])
                        y = float(line[38:46])
                        z = float(line[46:54])
                        coords.append([x, y, z])
                        break
    return np.array(coords)

# Function to write new PDB file with transformed coordinates
def write_pdb(pdb_file, output_file, transformation, center_of_mass):
    with open(pdb_file, 'r') as file, open(output_file, 'w') as out_file:
        for line in file:
            if line.startswith("ATOM"):
                x = float(line[30:38])
                y = float(line[38:46])
                z = float(line[46:54])
                coords = np.array([x, y, z])
                adjusted_coords = coords - center_of_mass
                transformed_coords = transformation @ adjusted_coords
                out_file.write(f"{line[:30]:<30}{transformed_coords[0]:8.3f}{transformed_coords[1]:8.3f}{transformed_coords[2]:8.3f}{line[54:]}")
            else:
                out_file.write(line)

# Define input and output directories
input_dir = './rigid_body_transformations/input/'
output_dir = './rigid_body_transformations/rotation_1/'

# Input PDB files, their respective residue ranges, and alignment specifications
pdb_files = [
    # {
    #     'file': 'a7_c3.pdb',
    #     'residue_ranges': [('A', (2,17)), ('B', (77,92)), ('C', (104,109))],
    #     'eigenvalue_index': 2,
    #     'axis': 'z'
    # },
    # {
    #     'file': 'e4_c4.pdb',
    #     'residue_ranges': [('A', (2,2)), ('A', (9,9)), ('A', (13,13)), ('A', (20,20)), ('A', (54,54)), ('A', (62,62)), ('A', (65,65)), ('A', (72,72)), ('A', (106,106)), ('A', (113,113)), ('A', (117,117)), ('A', (124,124)), ('A', (158,158)), ('A', (165,165)), ('A', (169,169)), ('A', (176,176)), ('A', (210,210)), ('A', (217,217)), ('A', (221,221)), ('A', (228,228)), ('A', (262,262)), ('A', (269,269)), ('A', (273,273)), ('A', (280,280))],
    #     'eigenvalue_index': 2,
    #     'axis': 'x'
    # },
    # {
    #     'file': 'f10_c6.pdb',
    #     'residue_ranges': [('A', (2,2)), ('A', (9,9)), ('A', (13,13)), ('A', (20,20)), ('A', (54,54)), ('A', (62,62)), ('A', (65,65)), ('A', (72,72)), ('A', (106,106)), ('A', (113,113)), ('A', (117,117)), ('A', (124,124)), ('A', (158,158)), ('A', (165,165)), ('A', (169,169)), ('A', (176,176)), ('A', (210,210)), ('A', (217,217)), ('A', (221,221)), ('A', (228,228)), ('A', (262,262)), ('A', (269,269)), ('A', (273,273)), ('A', (280,280))],
    #     'eigenvalue_index': 2,
    #     'axis': 'x'
    # }
    # {
    #     'file': 'c2.pdb',
    #     'residue_ranges': [('A', (3,3)), ('A', (73,73)), ('B', (3,3)), ('B', (73,73))],
    #     'eigenvalue_index': 1,
    #     'axis': 'y'
    # },
    # {
    #     'file': 'c4.pdb',
    #     'residue_ranges': [('A', (212,288)), ('C', (212,288))],
    #     'eigenvalue_index': 0,
    #     'axis': 'y'
    # },
        {
        'file': 'array_test.pdb',
        'residue_ranges': [('A', (1,500)), ('B', (1,500)), ('C', (1,500)), ('D', (1,500)), ('E', (1,500)), ('F', (1,500))],
        'eigenvalue_index': 0,
        'axis': 'x'
    },
    #     {
    #     'file': 'c3_fd31.pdb',
    #     'residue_ranges': [('A', (1,336)), ('B', (1,336)), ('C', (1,336))],
    #     'eigenvalue_index': 2,
    #     'axis': 'z'
    # },
]

# Define target axes
axis_map = {'x': [1, 0, 0], 'y': [0, 1, 0], 'z': [0, 0, 1]}

for pdb_spec in pdb_files:
    pdb_file = os.path.join(input_dir, pdb_spec['file'])
    residue_ranges = pdb_spec['residue_ranges']
    eigenvalue_index = pdb_spec['eigenvalue_index']
    target_axis = axis_map[pdb_spec['axis']]

    # Extract coordinates for the specified residues
    coords = parse_pdb(pdb_file, residue_ranges)

    # Calculate center of mass
    center_of_mass = np.mean(coords, axis=0)

    # Subtract center of mass from all coordinates
    adjusted_coords = coords - center_of_mass

    # Perform PCA to get eigenvectors
    cov_matrix = np.cov(adjusted_coords, rowvar=False)
    eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

    # Sort eigenvalues and eigenvectors
    sorted_indices = np.argsort(eigenvalues)[::-1]
    sorted_eigenvectors = eigenvectors[:, sorted_indices]

    # Select the specified eigenvector
    selected_eigenvector = sorted_eigenvectors[:, eigenvalue_index]

    # Align the selected eigenvector to the specified axis
    rotation = R.align_vectors([target_axis], [selected_eigenvector])[0]
    rotation_matrix = rotation.as_matrix()

    # Write the new PDB file with the transformed coordinates
    output_pdb_file = os.path.join(output_dir, pdb_spec['file'])
    write_pdb(pdb_file, output_pdb_file, rotation_matrix, center_of_mass)

    print(f"New PDB file with aligned structure saved to: {output_pdb_file}")
```

```python
import os
import numpy as np
from Bio.PDB import PDBParser, PDBIO, Chain
from Bio.PDB.vectors import rotaxis, Vector

input_dir = './rigid_body_transformations/translate_2/'
output_dir = './rigid_body_transformations/2d_array/'

def rotate_structure(structure, axis, angle):
    """Rotate a structure around a given axis by a given angle."""
    rotation_matrix = rotaxis(angle, Vector(axis))
    for model in structure:
        for chain in model:
            for residue in chain:
                for atom in residue:
                    atom.transform(rotation_matrix, np.zeros(3))
    return structure

def generate_cyclic_symmetry(pdb_file, symmetry_order, axis):
    """Generate C symmetry for a given PDB structure."""
    parser = PDBParser(QUIET=True)
    structure = parser.get_structure(pdb_file, pdb_file)
    angle = 2 * np.pi / symmetry_order
    rotated_structures = [structure]

    for i in range(1, symmetry_order):
        new_structure = structure.copy()
        rotate_structure(new_structure, axis, angle * i)
        rotated_structures.append(new_structure)

    return rotated_structures

def combine_structures(rotated_structures):
    """Combine rotated structures into a single structure with unique chain IDs."""
    combined_structure = rotated_structures[0].copy()
    model = combined_structure[0]
    existing_chain_ids = {chain.id for chain in model}
    
    for i, structure in enumerate(rotated_structures):
        if i == 0:
            continue
        for chain in structure[0]:
            new_id = chr(ord('A') + i)
            while new_id in existing_chain_ids:
                new_id = chr(ord(new_id) + 1)
            chain.id = new_id
            existing_chain_ids.add(new_id)
            model.add(chain)
    
    return combined_structure

def save_structure(structure, filename):
    """Save the structure to a PDB file."""
    io = PDBIO()
    io.set_structure(structure)
    io.save(filename)

def process_pdb_files(input_dir, output_dir, symmetry_dict, axis):
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
        
    pdb_files = [f for f in os.listdir(input_dir) if f.endswith('.pdb')]

    for pdb_file in pdb_files:
        if pdb_file in symmetry_dict:
            symmetry_order = symmetry_dict[pdb_file]
            file_path = os.path.join(input_dir, pdb_file)
            rotated_structures = generate_cyclic_symmetry(file_path, symmetry_order, axis)
            combined_structure = combine_structures(rotated_structures)
            
            output_file_path = os.path.join(output_dir, f"{os.path.splitext(pdb_file)[0]}_C{symmetry_order}.pdb")
            save_structure(combined_structure, output_file_path)

if __name__ == "__main__":
    # Specify the symmetry order for each PDB file
    symmetry_dict = {
        'fd31_r6_pk_model_5_ptm_seed_0_unrelaxed_y14_C3_y-59.pdb': 2,
        'fd31_r6_pk_model_5_ptm_seed_0_unrelaxed_y22_C4_y-65.pdb': 2,
        'fd31_r6_pk_model_5_ptm_seed_0_unrelaxed_y37_C6_y-80.pdb': 2,
        'fd31_r6_ty_model_5_ptm_seed_0_unrelaxed_y14_C3_y-59.pdb': 2,
        'fd31_r6_ty_model_5_ptm_seed_0_unrelaxed_y22_C4_y-65.pdb': 2,
        'fd31_r6_ty_model_5_ptm_seed_0_unrelaxed_y37_C6_y-80.pdb': 2,
    }
    
    axis = np.array([1, 0, 0])  # Rotate about the z-axis

    process_pdb_files(input_dir, output_dir, symmetry_dict, axis)
```

```python
import os
import numpy as np
from Bio import PDB

def calculate_chain_distance(chain):
    distances = []
    for residue in chain:
        for atom in residue:
            distances.append(atom.coord)
    mean_distance = np.mean(np.linalg.norm(distances, axis=1))
    return mean_distance

def filter_pdb_chains(input_dir):
    parser = PDB.PDBParser(QUIET=True)
    io = PDB.PDBIO()

    for filename in os.listdir(input_dir):
        if filename.endswith(".pdb"):
            pdb_path = os.path.join(input_dir, filename)
            structure = parser.get_structure(filename, pdb_path)
            
            chain_distances = []
            for model in structure:
                for chain in model:
                    distance = calculate_chain_distance(chain)
                    chain_distances.append((chain, distance))
            
            chain_distances.sort(key=lambda x: x[1])
            selected_chains = [chain for chain, _ in chain_distances[:6]]
            
            new_structure = PDB.Structure.Structure(structure.id)
            new_model = PDB.Model.Model(model.id)
            new_structure.add(new_model)
            for chain in selected_chains:
                new_model.add(chain)
            
            io.set_structure(new_structure)
            io.save(pdb_path)

if __name__ == "__main__":
    input_dir = "./diffusion/input"
    filter_pdb_chains(input_dir)
```

```python
import os
import pandas as pd
import numpy as np
from Bio.PDB import PDBParser, Superimposer
import shutil

# Define directories
packed_dir = './mpnn/outputs/packed'
superfold_outputs_dir = './superfold/outputs'
rmsd_filtered_dir = './superfold/rmsd_filtered'

# Create the filtered directory if it doesn't exist
os.makedirs(rmsd_filtered_dir, exist_ok=True)

# Initialize PDB parser
parser = PDBParser()

def calculate_rmsd(file1, file2):
    """Calculate RMSD between two PDB files."""
    structure1 = parser.get_structure('structure1', file1)
    structure2 = parser.get_structure('structure2', file2)
    
    # Extract atoms
    atoms1 = list(structure1.get_atoms())
    atoms2 = list(structure2.get_atoms())
    
    # Superimpose structures
    super_imposer = Superimposer()
    super_imposer.set_atoms(atoms1, atoms2)
    
    return super_imposer.rms

# Dataframe to store paths and RMSD values
data = []

# Iterate over files in the packed directory
for packed_pdb in os.listdir(packed_dir):
    if packed_pdb.endswith('.pdb'):
        # print(packed_pdb)
        # Construct corresponding file names in superfold directory
        superfold_pdb = packed_pdb.replace('.pdb', '_model_3_ptm_seed_0_unrelaxed.pdb')
        print(superfold_pdb)
        superfold_json = superfold_pdb.replace('unrelaxed.pdb', 'prediction_results.json')
        
        superfold_pdb_path = os.path.join(superfold_outputs_dir, superfold_pdb)
        superfold_json_path = os.path.join(superfold_outputs_dir, superfold_json)
        
        # Ensure the corresponding superfold pdb and json files exist
        if os.path.exists(superfold_pdb_path) and os.path.exists(superfold_json_path):
            packed_pdb_path = os.path.join(packed_dir, packed_pdb)
            
            # Calculate RMSD
            rmsd = calculate_rmsd(packed_pdb_path, superfold_pdb_path)
            
            # Check RMSD threshold
            if rmsd < 2:
                # Copy pdb and json files to rmsd_filtered directory
                shutil.copy(superfold_pdb_path, rmsd_filtered_dir)
                shutil.copy(superfold_json_path, rmsd_filtered_dir)
                
                # Add to dataframe
                data.append([packed_pdb_path, superfold_pdb_path, rmsd])

# Convert to DataFrame and save to CSV
df = pd.DataFrame(data, columns=['Packed PDB Path', 'Superfold PDB Path', 'RMSD'])
csv_path = os.path.join(rmsd_filtered_dir, 'rmsd_filtered.csv')
df.to_csv(csv_path, index=False)

print(f'RMSD calculation and file filtering complete. Results saved to {csv_path}')
```

```python
import os
import hashlib
from Bio import SeqIO
from Bio.PDB import PDBParser, Polypeptide
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord

input_dir = './superfold/rmsd_filtered'
output_dir = './superfold/fasta'

# Create output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

def get_chain_a_sequence(pdb_file, tile_type):
    parser = PDBParser(QUIET=True)
    structure = parser.get_structure('structure', pdb_file)
    
    chain_a_seq = ''
    for model in structure:
        for chain in model:
            if chain.id == 'A':
                for residue in chain:
                    if residue.get_id()[0] == ' ':
                        chain_a_seq += Polypeptide.three_to_one(residue.resname)
    
    if tile_type == 'tile':
        # Remove the first 52 residues
        return chain_a_seq[52:]
    else:
        # Remove the first 27 residues
        return chain_a_seq[27:]

def hash_filename(filename):
    return hashlib.md5(filename.encode()).hexdigest()

def pdb_to_fasta(pdb_file_1, pdb_file_2, fasta_file, suffix):
    seq_1 = get_chain_a_sequence(pdb_file_1, 'non-tile')
    seq_2 = get_chain_a_sequence(pdb_file_2, 'tile')
    
    final_sequence = seq_1 + seq_2
    
    seq = Seq(final_sequence)
    record = SeqRecord(seq, id=os.path.basename(pdb_file_1)[53:-100] + os.path.basename(pdb_file_1)[107:-43] + os.path.basename(pdb_file_2)[100:-55] + os.path.basename(pdb_file_2)[114:-44], description=f"")
    
    # Print the record
    print(record)
    
    # Write to fasta file
    SeqIO.write(record, fasta_file, 'fasta')

# Collect files
tile_a_files = [f for f in os.listdir(input_dir) if f.endswith('.pdb') and 'tile_a' in f]
tile_b_files = [f for f in os.listdir(input_dir) if f.endswith('.pdb') and 'tile_b' in f]
non_tile_files = [f for f in os.listdir(input_dir) if f.endswith('.pdb') and 'tile' not in f]

# Process tile_a + non-tile pairs
for tile_file in tile_a_files:
    for non_tile_file in non_tile_files:
        pdb_file_path_tile = os.path.join(input_dir, tile_file)
        pdb_file_path_non_tile = os.path.join(input_dir, non_tile_file)
        fasta_file_path = os.path.join(output_dir, hash_filename(tile_file) + '_' + hash_filename(non_tile_file) + '_tile_a.fasta')
        
        # Print the pair
        print(f"Processing pair: {tile_file} + {non_tile_file}")
        
        pdb_to_fasta(pdb_file_path_non_tile, pdb_file_path_tile, fasta_file_path, '_tile_a')

# Process tile_b + non-tile pairs
for tile_file in tile_b_files:
    for non_tile_file in non_tile_files:
        pdb_file_path_tile = os.path.join(input_dir, tile_file)
        pdb_file_path_non_tile = os.path.join(input_dir, non_tile_file)
        fasta_file_path = os.path.join(output_dir, hash_filename(tile_file) + '_' + hash_filename(non_tile_file) + '_tile_b.fasta')
        
        # Print the pair
        print(f"Processing pair: {tile_file} + {non_tile_file}")
        
        pdb_to_fasta(pdb_file_path_non_tile, pdb_file_path_tile, fasta_file_path, '_tile_b')

print("Conversion completed. FASTA files are located in:", output_dir)
```

```python
import os
from Bio import PDB
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO

def extract_sequences_from_pdb(pdb_file):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure('structure', pdb_file)
    
    sequences = []
    for model in structure:
        for chain in model:
            seq = []
            for residue in chain:
                if PDB.is_aa(residue):
                    seq.append(PDB.Polypeptide.three_to_one(residue.resname))
            sequence = ''.join(seq)
            chain_id = chain.id
            sequences.append((chain_id, sequence))
    
    return sequences

def write_fasta(sequences, output_file):
    records = []
    for pdb_id, chain_id, sequence in sequences:
        record = SeqRecord(Seq(sequence), id=f"{pdb_id}_{chain_id}", description="")
        records.append(record)
    SeqIO.write(records, output_file, "fasta")

def main():
    input_dir = './superfold/array_outputs'
    output_file = './superfold/array_outputs/sequences.fasta'
    
    all_sequences = []
    
    for filename in os.listdir(input_dir):
        if filename.endswith('.pdb'):
            pdb_file = os.path.join(input_dir, filename)
            pdb_id = os.path.splitext(filename)[0]
            sequences = extract_sequences_from_pdb(pdb_file)
            for chain_id, sequence in sequences:
                all_sequences.append((pdb_id, chain_id, sequence))
    
    write_fasta(all_sequences, output_file)
    print(f"Sequences written to {output_file}")

if __name__ == "__main__":
    main()
```

```python
import json
from Bio import SeqIO

# Define the input and output paths
input_fasta_path = './superfold/array_outputs/sequences.fasta'
output_json_path = './output_sequences.json'

# Read the sequences from the FASTA file using Biopython
def read_fasta(file_path):
    sequences = []
    for record in SeqIO.parse(file_path, "fasta"):
        sequences.append({'name': record.id, 'sequence': str(record.seq)})
    return sequences

# Determine the count based on the name
def determine_count(name):
    if "C6" in name:
        return 3
    elif "C4" in name:
        return 4
    elif "C3" in name:
        return 3
    else:
        return 1

# Generate the JSON structure
def generate_json(sequences):
    jobs = [
        {
            "name": seq['name'],
            "modelSeeds": [],
            "sequences": [
                {
                    "proteinChain": {
                        "sequence": seq['sequence'],
                        "glycans": [],
                        "modifications": [],
                        "count": determine_count(seq['name'])
                    }
                }
            ]
        } for seq in sequences
    ]
    return jobs

# Read sequences from the input FASTA file
sequences = read_fasta(input_fasta_path)

# Generate the JSON structure
json_data = generate_json(sequences)

# Write the JSON data to the output file
with open(output_json_path, 'w') as f:
    json.dump(json_data, f, indent=2)

print(f"JSON file has been created at {output_json_path}")
```
