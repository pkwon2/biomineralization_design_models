```python
#!/usr/bin/env python3

from pathlib import Path
import string

input_pdb = Path("/home/pkwon2/2d_array/array_propagation.pdb")
output_pdb = Path("/home/pkwon2/2d_array/array_propagation_relabelled.pdb")

# Allowed single-character PDB chain IDs
chain_ids = list(string.ascii_uppercase + string.ascii_lowercase + string.digits)

current_assigned_chain_index = -1
previous_input_chain = None

with input_pdb.open("r") as fin, output_pdb.open("w") as fout:
    for line in fin:
        record = line[:6].strip()

        if record in {"ATOM", "HETATM", "ANISOU", "TER"} and len(line) >= 22:
            input_chain = line[21]

            # Start a new output chain whenever the input chain ID changes
            if previous_input_chain is None or input_chain != previous_input_chain:
                current_assigned_chain_index += 1
                if current_assigned_chain_index >= len(chain_ids):
                    raise ValueError("Ran out of valid single-character chain IDs.")
                previous_input_chain = input_chain

            new_chain = chain_ids[current_assigned_chain_index]
            line = line[:21] + new_chain + line[22:]

        fout.write(line)

print(f"Wrote relabelled PDB to: {output_pdb}")
```
