```python
#!/usr/bin/env python3

from Bio import SeqIO
from Bio.SeqUtils import molecular_weight

fasta_file = "/home/pkwon2/aragonite/twist_order/cowboy/2025_08_22_2025_08_22_aragonite_n_array_e_coli_BsaI_into_LM0627.fa"

for record in SeqIO.parse(fasta_file, "fasta"):
    seq = str(record.seq).replace("*", "")  # remove stop codons
    mw = molecular_weight(seq, seq_type="protein")  
    print(f"{record.id[8:11]} {mw:.2f} Da")
```
