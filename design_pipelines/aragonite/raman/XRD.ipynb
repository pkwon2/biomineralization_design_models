```python
import numpy as np
import matplotlib.pyplot as plt

# File path
file_path = "/home/pkwon2/aragonite/raman/f8.xy"

# Load data: two columns separated by whitespace
data = np.loadtxt(file_path)

# Split into x and y
x = data[:, 0]
y = data[:, 1]

# ----- Restrict analysis to x >= 25 -----
mask = x >= 25
x = x[mask]
y = y[mask]

# Basic analysis
print(f"Number of data points: {len(x)}")
print(f"x range (analyzed): {x.min():.2f} – {x.max():.2f}")
print(f"y range (analyzed): {y.min():.2f} – {y.max():.2f}")

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="XRD spectrum (x ≥ 25°)")

# Labels and title
plt.xlabel("2θ (°)")
plt.ylabel("Intensity (a.u.)")
plt.title("XRD spectrum: ligandMPNN FD31")
plt.legend()

# --- Tick marks every 5° ---
xmin, xmax = x.min(), x.max()
plt.xticks(np.arange(np.floor(xmin / 5) * 5, np.ceil(xmax / 5) * 5 + 1, 5))

# Optional: minor ticks every 1°
plt.minorticks_on()
plt.tick_params(axis="x", which="minor", length=3)

plt.tight_layout()
plt.show()
```
