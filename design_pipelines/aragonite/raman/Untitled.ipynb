```python
import numpy as np
import matplotlib.pyplot as plt

# Path to your file
file_path = "/home/pkwon2/aragonite/raman/55C_buffer_5mM_caco3_pH_8.txt"

# Load the data
data = np.loadtxt(file_path)

# Extract columns
wave = data[:, 0]
intensity = data[:, 1]

# Plot
plt.figure(figsize=(10, 6))
plt.plot(wave, intensity, color="blue", linewidth=1.2)

# Labels and title
plt.xlabel("Raman Shift (cm$^{-1}$)")
plt.ylabel("Intensity (a.u.)")
plt.title("Raman Spectrum: 55C buffer 5mM CaCO₃ pH 8")
plt.grid(True, linestyle="--", alpha=0.6)

# Keep x-axis low → high
# (No invert_xaxis() here)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# File path
file_path = "/home/pkwon2/aragonite/raman/f2.xy"

# Load data: two columns separated by whitespace
data = np.loadtxt(file_path)

# Split into x and y
x = data[:, 0]
y = data[:, 1]

# Basic analysis
print(f"Number of data points: {len(x)}")
print(f"x range: {x.min()} – {x.max()}")
print(f"y range: {y.min()} – {y.max()}")

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="XRD spectrum")

# Label axes
plt.xlabel("2θ (°)")  # use 2θ notation if this is an XRD pattern
plt.ylabel("Intensity (a.u.)")
plt.title("XRD spectrum: f2")
plt.legend()

# --- Add tick marks every 5 degrees on the x-axis ---
xmin, xmax = x.min(), x.max()
plt.xticks(np.arange(np.floor(xmin / 5) * 5, np.ceil(xmax / 5) * 5 + 1, 5))

# Optional: add minor ticks every 1 degree for finer grid
plt.minorticks_on()
plt.tick_params(axis="x", which="minor", length=3)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# File path
file_path = "/home/pkwon2/aragonite/raman/f5.xy"

# Load data: two columns separated by whitespace
data = np.loadtxt(file_path)

# Split into x and y
x = data[:, 0]
y = data[:, 1]

# ----- Restrict analysis to x >= 25 -----
mask = x >= 25
x = x[mask]
y = y[mask]

# Basic analysis
print(f"Number of data points: {len(x)}")
print(f"x range (analyzed): {x.min():.2f} – {x.max():.2f}")
print(f"y range (analyzed): {y.min():.2f} – {y.max():.2f}")

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, label="XRD spectrum (x ≥ 25°)")

# Labels and title
plt.xlabel("2θ (°)")
plt.ylabel("Intensity (a.u.)")
plt.title("XRD spectrum: f5 (≥25°)")
plt.legend()

# --- Tick marks every 5° ---
xmin, xmax = x.min(), x.max()
plt.xticks(np.arange(np.floor(xmin / 5) * 5, np.ceil(xmax / 5) * 5 + 1, 5))

# Optional: minor ticks every 1°
plt.minorticks_on()
plt.tick_params(axis="x", which="minor", length=3)

plt.tight_layout()
plt.show()
```
