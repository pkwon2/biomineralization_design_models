```python
#!/usr/bin/env python3
import os
from Bio import PDB

def find_non_charged_residues(file_path):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure('PDB_structure', file_path)
    residue_positions = []
    charged_residues = ['GLU']

    for model in structure:
        for chain in model:
            residues_list = list(chain.get_residues())
            if len(residues_list) > 1:
                residues_to_process = residues_list[:]
            else:
                residues_to_process = []

            for residue in residues_to_process:
                resname = residue.get_resname()
                residue_number = residue.id[1]

                if resname not in charged_residues:
                    residue_positions.append(f"{chain.id}{residue_number}")
    return ' '.join(residue_positions)

def find_fixed_residues(file_path):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure('PDB_structure', file_path)
    fixed_positions = []

    for model in structure:
        for chain in model:
            for residue in chain.get_residues():
                if residue.get_resname() == 'GLU':
                    residue_number = residue.id[1]
                    fixed_positions.append(f"{chain.id}{residue_number}")
    return ' '.join(fixed_positions)

def main():
    pdb_dir = './mpnn'
    pdb_dir = os.path.abspath(pdb_dir)

    if not os.path.exists(pdb_dir):
        print(f"Directory {pdb_dir} does not exist.")
        return

    base_directory = 'mpnn'
    os.makedirs(base_directory, exist_ok=True)

    base_command = (
        "/software/containers/mlfold.sif /databases/mpnn/fused_mpnn/run.py "
        "--model_type 'ligand_mpnn' "
        "--temperature 0.3 "
        "--redesigned_residues '{residues}' "
        "--ligand_mpnn_use_side_chain_context 0 "
        "--fixed_residues '{fixed_residues}' "
        "--pdb_path {file_path} "
        "--out_folder outputs "
        "--batch_size 1 "
        "--number_of_batches 25 "
        "--bias_AA 'M:-10.0' "
        "--bias_AA 'C:-100.0' "
        "--pack_side_chains 1 "
        "--repack_everything 0"
    )

    tasks_file = os.path.join(base_directory, 'tasks')

    with open(tasks_file, 'w') as file:
        for pdb_file in os.listdir(pdb_dir):
            if pdb_file.endswith('.pdb'):
                file_path = os.path.join(pdb_dir, pdb_file)
                residues = find_non_charged_residues(file_path)
                fixed_residues = find_fixed_residues(file_path)
                command = base_command.format(
                    file_path=file_path,
                    residues=residues,
                    fixed_residues=fixed_residues
                )
                file.write(f"{command}\n")

    print(f"Tasks file '{tasks_file}' has been created.")

if __name__ == '__main__':
    main()
```

```python
#!/usr/bin/env python3
import os
from Bio import PDB
from Bio.PDB.Polypeptide import is_aa

def residue_tag(residue, chain_id):
    """
    Return a position tag like 'A123' or 'A123A' if an insertion code is present.
    Biopython residue.id is a tuple: (hetflag, resseq, icode)
    """
    hetflag, resseq, icode = residue.id
    tag = f"{chain_id}{resseq}"
    if icode and icode != " ":
        tag += icode  # preserve insertion code (e.g., A, B, etc.)
    return tag

def find_gly_residues(file_path):
    """Return space-separated chain/number tags for GLY residues (these will be redesignable)."""
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure('PDB_structure', file_path)
    redesign_positions = []

    for model in structure:
        for chain in model:
            for residue in chain.get_residues():
                # Skip hetero residues (waters, ligands) and keep only standard amino acids
                if residue.id[0] != " " or not is_aa(residue, standard=True):
                    continue
                if residue.get_resname().upper() == "GLY":
                    redesign_positions.append(residue_tag(residue, chain.id))

    return " ".join(redesign_positions)

def find_fixed_residues_except_gly(file_path):
    """Return space-separated chain/number tags for all standard residues except GLY (fixed)."""
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure('PDB_structure', file_path)
    fixed_positions = []

    for model in structure:
        for chain in model:
            for residue in chain.get_residues():
                if residue.id[0] != " " or not is_aa(residue, standard=True):
                    continue
                if residue.get_resname().upper() != "GLY":
                    fixed_positions.append(residue_tag(residue, chain.id))

    return " ".join(fixed_positions)

def main():
    # Input PDBs now come from ./mpnn/input
    pdb_dir = os.path.abspath('./mpnn/input')

    if not os.path.exists(pdb_dir):
        print(f"Directory {pdb_dir} does not exist.")
        return

    base_directory = 'mpnn'
    os.makedirs(base_directory, exist_ok=True)

    base_command = (
        "/software/containers/mlfold.sif /databases/mpnn/fused_mpnn/run.py "
        "--temperature 0.3 "
        "--homo_oligomer 1 "
        "--redesigned_residues '{residues}' "
        "--fixed_residues '{fixed_residues}' "
        "--pdb_path {file_path} "
        "--out_folder outputs "
        "--batch_size 4 "
        "--number_of_batches 4 "
        "--bias_AA 'M:-10.0' "
        "--pack_side_chains 1 "
        "--repack_everything 1"
    )

    tasks_file = os.path.join(base_directory, 'tasks')

    with open(tasks_file, 'w') as file:
        for pdb_file in sorted(os.listdir(pdb_dir)):
            if pdb_file.lower().endswith('.pdb'):
                file_path = os.path.join(pdb_dir, pdb_file)

                # Redesign only glycines
                residues = find_gly_residues(file_path)

                # Fix everything else
                fixed_residues = find_fixed_residues_except_gly(file_path)

                command = base_command.format(
                    file_path=file_path,
                    residues=residues,
                    fixed_residues=fixed_residues
                )
                file.write(f"{command}\n")

    print(f"Tasks file '{tasks_file}' has been created, using inputs from ./mpnn/input.")

if __name__ == '__main__':
    main()
```

```python
#!/usr/bin/env python3
import os
from Bio import PDB

def find_non_charged_residues(file_path):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure('PDB_structure', file_path)
    residue_positions = []
    charged_residues = ['']

    for model in structure:
        for chain in model:
            residues_list = list(chain.get_residues())
            if len(residues_list) > 1:
                residues_to_process = residues_list[:]
            else:
                residues_to_process = []

            for residue in residues_to_process:
                resname = residue.get_resname()
                residue_number = residue.id[1]

                if resname not in charged_residues:
                    residue_positions.append(f"{chain.id}{residue_number}")
    return ' '.join(residue_positions)

def find_fixed_residues(file_path):
    parser = PDB.PDBParser(QUIET=True)
    structure = parser.get_structure('PDB_structure', file_path)
    fixed_positions = []

    for model in structure:
        for chain in model:
            for residue in chain.get_residues():
                if residue.get_resname() == '':
                    residue_number = residue.id[1]
                    fixed_positions.append(f"{chain.id}{residue_number}")
    return ' '.join(fixed_positions)

def main():
    pdb_dir = './mpnn'
    pdb_dir = os.path.abspath(pdb_dir)

    if not os.path.exists(pdb_dir):
        print(f"Directory {pdb_dir} does not exist.")
        return

    base_directory = 'mpnn'
    os.makedirs(base_directory, exist_ok=True)

    base_command = (
        "/software/containers/mlfold.sif /databases/mpnn/fused_mpnn/run.py "
        "--model_type 'ligand_mpnn' "
        "--temperature 0.3 "
        "--redesigned_residues '{residues}' "
        "--ligand_mpnn_use_side_chain_context 1 "
        "--fixed_residues '{fixed_residues}' "
        "--pdb_path {file_path} "
        "--out_folder side_chains_unfixed_outputs "
        "--batch_size 2 "
        "--number_of_batches 2 "
        "--bias_AA 'M:-10.0' "
        "--bias_AA 'C:-100.0' "
        "--pack_side_chains 1 "
        "--repack_everything 0"
    )

    tasks_file = os.path.join(base_directory, 'side_chains_unfixed_tasks')

    with open(tasks_file, 'w') as file:
        for pdb_file in os.listdir(pdb_dir):
            if pdb_file.endswith('.pdb'):
                file_path = os.path.join(pdb_dir, pdb_file)
                residues = find_non_charged_residues(file_path)
                fixed_residues = find_fixed_residues(file_path)
                command = base_command.format(
                    file_path=file_path,
                    residues=residues,
                    fixed_residues=fixed_residues
                )
                file.write(f"{command}\n")

    print(f"Tasks file '{tasks_file}' has been created.")

if __name__ == '__main__':
    main()
```

```python
import os

def keep_chain_a(input_file, output_file):
    """
    Keeps only chain A in a PDB file and writes the result to the output file.
    """
    with open(input_file, 'r') as f_in, open(output_file, 'w') as f_out:
        for line in f_in:
            # Keep ATOM/HETATM records from chain A
            if (line.startswith("ATOM") or line.startswith("HETATM")) and line[21] == 'A':
                f_out.write(line)
            # Keep headers/footers (TER, END, etc.)
            elif not (line.startswith("ATOM") or line.startswith("HETATM")):
                f_out.write(line)

def process_pdb_files(input_directory, output_directory, suffix):
    """
    Processes all .pdb files in the specified directory, keeping only chain A.
    Writes the modified files to the output directory with the given suffix.
    """
    os.makedirs(output_directory, exist_ok=True)
    
    for filename in os.listdir(input_directory):
        if filename.endswith(".pdb"):
            input_file = os.path.join(input_directory, filename)
            
            # Strip existing .pdb and add suffix
            base_name = os.path.splitext(filename)[0]
            output_file = os.path.join(output_directory, base_name + suffix)
            
            keep_chain_a(input_file, output_file)
            print(f"Processed {filename} -> {os.path.basename(output_file)}")

if __name__ == "__main__":
    # Define input/output directories
    fixed_input = './mpnn/outputs/packed'
    unfixed_input = './mpnn/side_chains_unfixed_outputs/packed'
    output_directory = './mpnn/outputs/alphafold'
    
    # Process fixed -> add _fixed.pdb suffix
    process_pdb_files(fixed_input, output_directory, "_fixed.pdb")
    
    # Process unfixed -> add _unfixed.pdb suffix
    process_pdb_files(unfixed_input, output_directory, "_unfixed.pdb")
```

```python
base_directory = "superfold"
create_directory_structure(base_directory)

import os
import pandas as pd
import numpy as np
from Bio.PDB import PDBParser, Superimposer
import shutil

# Define directories
packed_dir = './mpnn/outputs/alphafold'
superfold_outputs_dir = './superfold/outputs'
rmsd_filtered_dir = './superfold/rmsd_filtered'

# Create the filtered directory if it doesn't exist
os.makedirs(rmsd_filtered_dir, exist_ok=True)

# Initialize PDB parser
parser = PDBParser(QUIET=True)

def calculate_rmsd(file1, file2):
    """Calculate RMSD between two PDB files using CA atoms."""
    try:
        structure1 = parser.get_structure('structure1', file1)
        structure2 = parser.get_structure('structure2', file2)

        # Extract CA atoms and order by residue number
        atoms1 = []
        atoms2 = []
        for model1, model2 in zip(structure1, structure2):
            for chain1, chain2 in zip(model1, model2):
                residues1 = [res for res in chain1 if res.has_id('CA')]
                residues2 = [res for res in chain2 if res.has_id('CA')]

                if len(residues1) != len(residues2):
                    print(f'Number of CA residues differ between {file1} and {file2}')
                    return None

                for res1, res2 in zip(residues1, residues2):
                    atoms1.append(res1['CA'])
                    atoms2.append(res2['CA'])

        if len(atoms1) != len(atoms2):
            print(f'Number of CA atoms differ between {file1} and {file2}')
            return None

        # Superimpose structures
        super_imposer = Superimposer()
        super_imposer.set_atoms(atoms1, atoms2)
        super_imposer.apply(structure2.get_atoms())

        return super_imposer.rms
    except Exception as e:
        print(f'Error calculating RMSD between {file1} and {file2}: {e}')
        return None

# Dataframe to store paths and RMSD values
data = []

# Iterate over files in the packed directory
for packed_pdb in os.listdir(packed_dir):
    if packed_pdb.endswith('.pdb'):
        packed_pdb_path = os.path.join(packed_dir, packed_pdb)
        # Loop over models 1 to 5
        for model_num in range(3, 4):
            # Construct corresponding file names in superfold directory
            superfold_pdb = packed_pdb.replace('.pdb', f'_model_{model_num}_ptm_seed_0_unrelaxed.pdb')
            print(f'Processing: {superfold_pdb}')
            superfold_json = superfold_pdb.replace('unrelaxed.pdb', 'prediction_results.json')
            
            superfold_pdb_path = os.path.join(superfold_outputs_dir, superfold_pdb)
            superfold_json_path = os.path.join(superfold_outputs_dir, superfold_json)
            
            # Ensure the corresponding superfold pdb and json files exist
            if os.path.exists(superfold_pdb_path) and os.path.exists(superfold_json_path):
                # Calculate RMSD
                rmsd = calculate_rmsd(packed_pdb_path, superfold_pdb_path)
                
                if rmsd is not None:
                    # Check RMSD threshold
                    if rmsd < 1.5:
                        # Copy pdb and json files to rmsd_filtered directory
                        shutil.copy(superfold_pdb_path, rmsd_filtered_dir)
                        shutil.copy(superfold_json_path, rmsd_filtered_dir)
                        
                        # Add to dataframe
                        data.append([packed_pdb_path, superfold_pdb_path, rmsd])
                else:
                    print(f'Skipping due to RMSD calculation error between {packed_pdb_path} and {superfold_pdb_path}')
            else:
                print(f'Missing files for {superfold_pdb_path} or {superfold_json_path}')

# Convert to DataFrame and save to CSV
df = pd.DataFrame(data, columns=['Packed PDB Path', 'Superfold PDB Path', 'RMSD'])
csv_path = os.path.join(rmsd_filtered_dir, 'rmsd_filtered.csv')
df.to_csv(csv_path, index=False)

print(f'RMSD calculation and file filtering complete. Results saved to {csv_path}')
```

```python
import os
import json
import shutil
from collections import defaultdict

# Source and destination directories
src_dir = './superfold/rmsd_filtered'
dst_root = './superfold/order'
os.makedirs(dst_root, exist_ok=True)

# Initialize a dict: group_prefix -> list of (mean_pae, json_path, filename)
file_groups = defaultdict(list)

# Function to extract the prefix (kept as you provided)
def get_prefix(filename):
    parts = filename.split('value1')
    if len(parts) > 1:
        return 'value1'.join(parts[:-1]) + 'value1'
    return filename

# Discover all result jsons and bucket by prefix
for filename in os.listdir(src_dir):
    if filename.endswith('prediction_results.json'):
        prefix = get_prefix(filename)
        json_path = os.path.join(src_dir, filename)
        try:
            with open(json_path, 'r') as jf:
                data = json.load(jf)
            mean_pae = data.get('mean_pae', 999.0)  # large default so it won't rank highly
        except Exception as e:
            print(f"Warning: failed to read/parse {json_path}: {e}")
            mean_pae = 999.0
        file_groups[prefix].append((mean_pae, json_path, filename))

def ensure_group_dir(prefix: str) -> str:
    """
    Make a per-group subdirectory under dst_root that's filesystem-safe.
    """
    safe = prefix.replace(os.sep, "_").replace(" ", "_")
    grp_dir = os.path.join(dst_root, safe)
    os.makedirs(grp_dir, exist_ok=True)
    return grp_dir

def copy_pair(json_path: str, filename: str, dest_dir: str):
    """
    Copy the json and its matching unrelaxed.pdb to dest_dir (if present).
    """
    # Copy JSON
    dst_json = os.path.join(dest_dir, filename)
    shutil.copy2(json_path, dst_json)
    print(f"Copied {json_path} -> {dst_json}")

    # Copy corresponding PDB if it exists
    pdb_filename = filename.replace('prediction_results.json', 'unrelaxed.pdb')
    pdb_path = os.path.join(src_dir, pdb_filename)
    if os.path.exists(pdb_path):
        dst_pdb = os.path.join(dest_dir, pdb_filename)
        shutil.copy2(pdb_path, dst_pdb)
        print(f"Copied {pdb_path} -> {dst_pdb}")
    else:
        print(f"Note: no matching PDB found for {filename} at {pdb_path}")

# Process each group
TOP_N = 1
for prefix, files in file_groups.items():
    # Split into fixed vs unfixed by filename substring
    fixed = [(mp, jp, fn) for (mp, jp, fn) in files if '_fixed_' in fn]
    unfixed = [(mp, jp, fn) for (mp, jp, fn) in files if '_unfixed_' in fn]

    # Sort ascending by mean_pae
    fixed.sort(key=lambda x: x[0])
    unfixed.sort(key=lambda x: x[0])

    # Take top N from each bucket
    pick_fixed = fixed[:TOP_N]
    pick_unfixed = unfixed[:TOP_N]

    # Where to copy
    group_dir = ensure_group_dir(prefix)

    # Copy selected files
    for mp, json_path, filename in (pick_fixed + pick_unfixed):
        copy_pair(json_path, filename, group_dir)

print("Ordering + copy complete.")
```

```python
import os
import re
import shutil
import pandas as pd

# ---- settings ----
base_directory = "superfold"
superfold_order_root = os.path.join(base_directory, "order")
dest_dir = os.path.join(base_directory, "paired")

# Packed search roots
packed_fixed_root = './mpnn/outputs/packed'
packed_unfixed_root = './mpnn/side_chains_unfixed_outputs/packed'

def create_directory_structure(base_dir: str):
    os.makedirs(os.path.join(base_dir, "outputs"), exist_ok=True)
    os.makedirs(os.path.join(base_dir, "pairedr"), exist_ok=True)

create_directory_structure(base_directory)
os.makedirs(dest_dir, exist_ok=True)

paired_records = []
missing_packed = []

def safe_dest(path: str) -> str:
    """If file exists in dest, add suffix to avoid overwrite."""
    if not os.path.exists(path):
        return path
    root, ext = os.path.splitext(path)
    i = 1
    new_path = f"{root}__{i}{ext}"
    while os.path.exists(new_path):
        i += 1
        new_path = f"{root}__{i}{ext}"
    return new_path

# Regex to strip fixed/unfixed suffix
SUFFIX_RE = re.compile(r'_(fixed|unfixed)_model_3_ptm_seed_0_unrelaxed\.pdb$')

def derive_packed_name(superfold_name: str) -> str:
    """
    Turn a superfold name like:
      foo_fixed_model_3_ptm_seed_0_unrelaxed.pdb
    into:
      foo.pdb
    """
    return SUFFIX_RE.sub('.pdb', superfold_name)

# Collect all superfold PDBs
superfold_items = []
if os.path.isdir(superfold_order_root):
    for group_name in os.listdir(superfold_order_root):
        group_dir = os.path.join(superfold_order_root, group_name)
        if not os.path.isdir(group_dir):
            continue
        for fname in os.listdir(group_dir):
            if fname.endswith('.pdb'):
                superfold_items.append((os.path.join(group_dir, fname), fname))

# Pair and copy
for superfold_pdb_path, superfold_pdb_name in superfold_items:
    if '_fixed_' in superfold_pdb_name:
        packed_root = packed_fixed_root
    elif '_unfixed_' in superfold_pdb_name:
        packed_root = packed_unfixed_root
    else:
        missing_packed.append((superfold_pdb_name, "no _fixed_/_unfixed_ tag"))
        continue

    packed_name = derive_packed_name(superfold_pdb_name)
    packed_pdb_path = os.path.join(packed_root, packed_name)

    if not os.path.exists(packed_pdb_path):
        missing_packed.append((superfold_pdb_name, packed_pdb_path))
        continue

    dest_superfold = safe_dest(os.path.join(dest_dir, superfold_pdb_name))
    dest_packed = safe_dest(os.path.join(dest_dir, packed_name))

    shutil.copy2(superfold_pdb_path, dest_superfold)
    shutil.copy2(packed_pdb_path, dest_packed)

    paired_records.append({
        "superfold_filename": os.path.basename(dest_superfold),
        "packed_filename": os.path.basename(dest_packed),
        "superfold_source": superfold_pdb_path,
        "packed_source": packed_pdb_path,
        "destination_dir": dest_dir
    })

# Write manifest
manifest_path = os.path.join(dest_dir, 'paired_copies_manifest.csv')
pd.DataFrame(paired_records).to_csv(manifest_path, index=False)

print(f"Pairing & copy complete. {len(paired_records)} pairs copied to {dest_dir}")

if missing_packed:
    print("\nMissing packed counterparts:")
    for super_name, expected_path in missing_packed[:20]:
        print(f"  superfold: {super_name}  |  expected packed at: {expected_path}")
    if len(missing_packed) > 20:
        print(f"  ... and {len(missing_packed) - 20} more")

print(f"\nManifest written to: {manifest_path}")
```

```python
#!/usr/bin/env python3
import os
import sys
import pandas as pd

INPUT = "/home/pkwon2/cobalt_carb/superfold/pairedr/example_plate-file-upload-2.xlsx"
# Writes a new .xlsx so your original stays intact
OUTPUT = os.path.splitext(INPUT)[0] + "_modified.xlsx"

# We’ll search for these likely column names and use the first one we find
CANDIDATE_COLS = ["Sequence", "sequence", "gene sequence", "Gene Sequence", "GeneSequence"]

def sniff_file_signature(path, n=8):
    with open(path, "rb") as f:
        head = f.read(n)
    return head

def load_table_anyway(path):
    """
    Try to load as true Excel first (XLS/XLSX by signature), then gracefully fall back to CSV/TSV.
    """
    head = sniff_file_signature(path)

    # XLSX/Excel Open XML files are ZIPs and start with 'PK'
    is_xlsx_like = head.startswith(b"PK")
    # Legacy XLS (OLE2/CFB) starts with D0 CF 11 E0 A1 B1 1A E1
    is_xls_like = head.startswith(b"\xD0\xCF\x11\xE0")

    # 1) Try XLSX with openpyxl if signature matches
    if is_xlsx_like:
        try:
            return pd.read_excel(path, engine="openpyxl")
        except Exception as e:
            print(f"[warn] openpyxl failed on a PK/zip-like file: {e}", file=sys.stderr)

    # 2) Try legacy XLS with xlrd (<=1.2.0 supports .xls)
    if is_xls_like or path.lower().endswith(".xls"):
        try:
            return pd.read_excel(path, engine="xlrd")
        except ImportError:
            raise SystemExit(
                "Missing xlrd for .xls. Install with:\n"
                "    pip install 'xlrd==1.2.0'\n"
                "Or re-save the file as .xlsx and try again."
            )
        except Exception as e:
            print(f"[warn] xlrd failed on an XLS-like file: {e}", file=sys.stderr)

    # 3) Fallbacks: it might actually be CSV/TSV (or HTML-exported CSV)
    # First try Python engine with sep=None (sniff delimiter)
    try:
        return pd.read_csv(path, sep=None, engine="python")
    except Exception as e:
        print(f"[warn] read_csv(sep=None) failed: {e}", file=sys.stderr)

    # Try tab-delimited explicitly
    try:
        return pd.read_csv(path, sep="\t", engine="python")
    except Exception as e:
        print(f"[warn] read_csv(tab) failed: {e}", file=sys.stderr)

    # Last resort: give a clear error
    raise SystemExit(
        "Could not open the file as XLSX (openpyxl), XLS (xlrd), or CSV/TSV. "
        "If this is an Excel export, re-save it as .xlsx and retry."
    )

def find_sequence_column(df, candidates):
    for c in candidates:
        if c in df.columns:
            return c
    # If nothing matched, show available columns
    raise SystemExit(
        "Could not find a gene/sequence column. Available columns:\n"
        + ", ".join(map(str, df.columns))
    )

def main():
    if not os.path.exists(INPUT):
        raise SystemExit(f"Input not found: {INPUT}")

    df = load_table_anyway(INPUT)

    seq_col = find_sequence_column(df, CANDIDATE_COLS)

    # Remove ALL occurrences of 'ATG' (case-insensitive). If you only want to strip
    # a leading start codon, replace the next line with:
    # df[seq_col] = df[seq_col].astype(str).str.replace(r"^(?i)ATG", "", regex=True)
    df[seq_col] = df[seq_col].astype(str).str.replace("ATG", "", case=False, regex=False)

    # Write clean .xlsx (openpyxl handles writing)
    df.to_excel(OUTPUT, index=False)
    print(f"Processed file saved to: {OUTPUT}")
    print(f"Column used: {seq_col}")

if __name__ == "__main__":
    main()
```

```python
import os
import pandas as pd
from Bio.PDB import PDBParser, Superimposer

# Paths
manifest_path = './superfold/6x_dhr_order/paired_copies_manifest.csv'  # or paired_moves_manifest.csv
output_csv = './superfold/6x_dhr_order/paired_rmsd_results.csv'

# Load manifest
df = pd.read_csv(manifest_path)

parser = PDBParser(QUIET=True)

def calculate_rmsd(file1, file2):
    """Calculate RMSD between two PDB files using CA atoms."""
    try:
        structure1 = parser.get_structure('s1', file1)
        structure2 = parser.get_structure('s2', file2)

        atoms1, atoms2 = [], []
        for model1, model2 in zip(structure1, structure2):
            for chain1, chain2 in zip(model1, model2):
                residues1 = [res for res in chain1 if res.has_id('CA')]
                residues2 = [res for res in chain2 if res.has_id('CA')]

                if len(residues1) != len(residues2):
                    print(f"Different number of CA residues in {file1} vs {file2}")
                    return None

                for r1, r2 in zip(residues1, residues2):
                    atoms1.append(r1['CA'])
                    atoms2.append(r2['CA'])

        if len(atoms1) != len(atoms2):
            print(f"Different number of CA atoms in {file1} vs {file2}")
            return None

        sup = Superimposer()
        sup.set_atoms(atoms1, atoms2)
        return sup.rms
    except Exception as e:
        print(f"Error calculating RMSD between {file1} and {file2}: {e}")
        return None

# Iterate over pairs and calculate RMSD
rmsd_records = []
for _, row in df.iterrows():
    packed_file = row['packed_source']
    superfold_file = row['superfold_source']

    if not (os.path.exists(packed_file) and os.path.exists(superfold_file)):
        print(f"Skipping: {packed_file} or {superfold_file} not found")
        continue

    rmsd_val = calculate_rmsd(packed_file, superfold_file)
    rmsd_records.append({
        "packed_file": packed_file,
        "superfold_file": superfold_file,
        "RMSD": rmsd_val
    })

# Save results
rmsd_df = pd.DataFrame(rmsd_records)
rmsd_df.to_csv(output_csv, index=False)

print(f"RMSD calculation complete. Results written to {output_csv}")
```

```python
import os
import glob
from Bio.PDB import PDBParser, PPBuilder
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO
from Bio.Seq import Seq

# Directory containing PDB files
glob_dir = './superfold/rmsd_filtered'
# Output FASTA file within pdb_dir
fasta_file = os.path.join(glob_dir, 'chainA.fasta')
# The head sequence to prepend before each chain A sequence
head_seq = ""

# Initialize the PDB parser, peptide builder, and list for sequence records
parser = PDBParser(QUIET=True)
ppb = PPBuilder()
records = []

# Loop over all .pdb files in the directory
for pdb_path in glob.glob(os.path.join(glob_dir, '*.pdb')):
    pdb_id = os.path.basename(pdb_path).split('.')[0]
    try:
        structure = parser.get_structure(pdb_id, pdb_path)
    except Exception as e:
        print(f"Error parsing {pdb_path}: {e}")
        continue

    model = structure[0]
    if 'A' not in model:
        print(f"Chain A not found in {pdb_path}")
        continue
    chain_a = model['A']

    # Build polypeptides (handles discontinuous segments)
    peptides = ppb.build_peptides(chain_a)
    seq_str = "".join(str(pp.get_sequence()) for pp in peptides)

    if not seq_str:
        print(f"No sequence extracted from chain A in {pdb_path}")
        continue

    # Exclude first and last 34 residues
    if len(seq_str) <= 68:
        print(f"Sequence from chain A in {pdb_id} shorter than 68 aa; skipping trimming.")
        continue
    trimmed_seq = seq_str[:]

    # Prepend head sequence and create FASTA record
    full_seq = head_seq + trimmed_seq
    record = SeqRecord(Seq(full_seq), id=f"{pdb_id}_A", description="")
    records.append(record)
    print(f"Extracted trimmed chain A from {pdb_id} (residues 35 to {len(seq_str)-34}).")

# Write all records to the FASTA file
if records:
    SeqIO.write(records, fasta_file, 'fasta')
    print(f"\nWrote {len(records)} record(s) to {fasta_file}")
else:
    print("No trimmed chain A sequences were extracted.")
```

```python
import os
from Bio import SeqIO

def create_directory_structure(base_directory):
    """
    Creates a subdirectory named 'fasta_files' within the base_directory.
    Returns the full path to the created subdirectory.
    """
    output_directory = os.path.join(base_directory, "fasta_files")
    if not os.path.exists(output_directory):
        os.makedirs(output_directory)
    return output_directory

def read_fasta(file_path):
    """
    Reads sequences from a FASTA file and returns a list of dictionaries,
    each containing the sequence name and its sequence string.
    """
    sequences = []
    for record in SeqIO.parse(file_path, "fasta"):
        sequences.append({'name': record.id, 'sequence': str(record.seq)})
    return sequences

def write_fasta_files(sequences, output_directory):
    """
    For each sequence, writes a separate FASTA file with:
    [sequence repeated 3x with no delimiter] + " " + [sequence repeated 3x with no delimiter]
    """
    for seq in sequences:
        repeated_seq = seq['sequence'] * 3
        fasta_content = f">{seq['name']}\n{repeated_seq} {repeated_seq}\n"
        fasta_file_path = os.path.join(output_directory, f"{seq['name']}.fasta")
        with open(fasta_file_path, 'w') as f:
            f.write(fasta_content)
        print(f"Created FASTA file: {fasta_file_path}")

if __name__ == "__main__":
    # Define base and output paths
    base_directory = "alphafold3"
    output_fasta_directory = create_directory_structure(base_directory)
    
    # Define the input FASTA file path
    input_fasta_path = './superfold/rmsd_filtered/chainA.fasta'
    
    # Read sequences from the FASTA file
    sequences = read_fasta(input_fasta_path)
    
    # Write modified FASTA files
    write_fasta_files(sequences, output_fasta_directory)
```

```python
#!/usr/bin/env python3
import os
import csv
import numpy as np
from glob import glob
from Bio.PDB.MMCIFParser import MMCIFParser

# Paths
in_dir  = "/home/pkwon2/aragonite/alphafold3/filtered"
out_csv = os.path.join(in_dir, "flatness_metrics.csv")

# Parser
parser = MMCIFParser(QUIET=True)

def get_calpha_coords(cif_file):
    """Extract C-alpha coordinates from an mmCIF file."""
    structure = parser.get_structure("struct", cif_file)
    coords = []
    for model in structure:
        for chain in model:
            for residue in chain:
                try:
                    atom = residue["CA"]  # dictionary-style access
                    coords.append(atom.coord)
                except KeyError:
                    continue
    return np.array(coords)

def compute_flatness(coords):
    """
    Compute flatness as variance of absolute distances from best-fit plane.
    Lower values = flatter.
    """
    if len(coords) < 3:
        return None
    # Center
    coords_centered = coords - coords.mean(axis=0)
    # PCA via SVD
    _, _, vh = np.linalg.svd(coords_centered)
    normal = vh[-1]  # normal to best-fit plane
    # Distances from plane
    dists = np.abs(coords_centered.dot(normal))
    # Use variance of absolute distances
    return np.var(dists)

def main():
    cif_files = glob(os.path.join(in_dir, "*_model_3_ptm_seed_0_unrelaxed_a_model.cif"))
    results = []

    for cif_fp in cif_files:
        coords = get_calpha_coords(cif_fp)
        flatness = compute_flatness(coords)
        base = os.path.basename(cif_fp)
        if flatness is not None:
            results.append((base, flatness))
            print(f"[OK] {base}: flatness_var = {flatness:.6f}")
        else:
            print(f"[SKIP] {base}: too few atoms")

    # Sort results by flatness (ascending = flattest first)
    results.sort(key=lambda x: x[1])

    # Write CSV
    with open(out_csv, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["filename", "flatness_variance"])
        writer.writerows(results)

    print(f"[DONE] Results written to {out_csv}")

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
"""
Linear-by-default propagation with conditional C6 snap that preserves the A/B plane.

Behavior:
- Default: use the estimated A->B rigid transform (iterative screw-like propagation).
- If A->B is near-linear (tiny angle) or near-180° flip: snap to a cyclic C6.
  * Define the "largest plane" from A and B by PCA (best-fit plane to A∪B).
  * Use the plane normal as the rotation axis.
  * Rotate 60° about a center lying in that plane.
  * Solve for the center so the A->B centroid shift projected in-plane is reproduced.
  * Remove any normal (out-of-plane) translation to keep the plane unchanged.

Inputs:  /home/pkwon2/aragonite/alphafold3/filtered/*.cif
Outputs: /home/pkwon2/aragonite/alphafold3/propagation/<stem>_propagated.pdb
"""

import os
import glob
import numpy as np
from math import acos, cos, sin, pi
from Bio.PDB import MMCIFParser, PDBIO
from Bio.PDB import Atom, Residue, Chain, Model, Structure
from Bio.PDB.Polypeptide import is_aa

INPUT_DIR = "/home/pkwon2/aragonite/twist_order/array_order"
OUTPUT_DIR = "/home/pkwon2/aragonite/twist_order/propagation"

CHAIN_FROM = "A"
CHAIN_TO   = "B"
OUTPUT_CHAIN_IDS = ["A", "B", "C", "D", "E", "F"]

# Snap only when rotation is almost linear/flip
ANGLE_MIN_DEG_FOR_SNAP = 15.0    # snap if angle < 15°
ANGLE_MAX_DEG_FOR_SNAP = 165.0   # or if angle > 165°

os.makedirs(OUTPUT_DIR, exist_ok=True)

def kabsch(P, Q):
    Pc = P.mean(axis=0); Qc = Q.mean(axis=0)
    P0, Q0 = P - Pc, Q - Qc
    H = P0.T @ Q0
    U, S, Vt = np.linalg.svd(H)
    R = Vt.T @ U.T
    if np.linalg.det(R) < 0.0:
        Vt[-1, :] *= -1.0
        R = Vt.T @ U.T
    t = Qc - R @ Pc
    return R, t

def rot_axis_angle(R):
    tr = np.clip((np.trace(R) - 1.0) / 2.0, -1.0, 1.0)
    theta = acos(tr)
    ux = R[2,1] - R[1,2]
    uy = R[0,2] - R[2,0]
    uz = R[1,0] - R[0,1]
    u = np.array([ux, uy, uz], dtype=float)
    n = np.linalg.norm(u)
    if n < 1e-8:
        w, v = np.linalg.eig(R)
        idx = np.argmin(np.abs(np.real(w) - 1.0))
        u = np.real(v[:, idx])
        n = np.linalg.norm(u)
        if n < 1e-8:
            u = np.array([0.0, 0.0, 1.0]); n = 1.0
    return u / n, theta

def R_from_axis_angle(u, theta):
    ux, uy, uz = u
    K = np.array([[0, -uz,  uy],
                  [uz,  0, -ux],
                  [-uy, ux,  0]], dtype=float)
    I = np.eye(3)
    return I + sin(theta) * K + (1 - cos(theta)) * (K @ K)

def get_first_model(structure):
    return next(structure.get_models(), None)

def get_chain_resolved(structure, target_id):
    model = get_first_model(structure)
    if model is None:
        return None, "no models found"
    chains = list(model.get_chains())
    if not chains:
        return None, "no chains in model"

    by_exact = {c.id: c for c in chains}
    if target_id in by_exact:
        return by_exact[target_id], f"exact '{target_id}'"

    target_norm = target_id.strip().upper()
    by_norm = {c.id.strip().upper(): c for c in chains}
    if target_norm in by_norm:
        return by_norm[target_norm], f"normalized '{target_norm}'"

    first_map = {}
    for c in chains:
        key = (c.id or "").strip().upper()[:1]
        if key and key not in first_map:
            first_map[key] = c
    if target_norm[:1] in first_map:
        c = first_map[target_norm[:1]]
        return c, f"first-char '{target_norm[:1]}' -> '{c.id}'"

    avail = ", ".join(sorted(c.id for c in chains))
    return None, f"could not resolve '{target_id}'. Available: {avail}"

def ca_map(chain):
    m = {}
    for res in chain:
        if not is_aa(res, standard=True):
            continue
        if "CA" in res:
            resseq = res.id[1]; icode = res.id[2]
            m[(resseq, icode)] = res["CA"].get_coord()
    return m

def extract_chain_template(chain):
    residues_template, coords = [], []
    for res in chain:
        if res.id[0] not in (" ", "A"):
            continue
        res_id = res.id
        resname = res.resname
        segid = getattr(res, "segid", "    ")
        atom_tuples = []
        for atom in res.get_atoms():
            if atom.get_parent().id[0] not in (" ", "A"):
                continue
            name = atom.get_name()
            fullname = atom.get_fullname()
            altloc = atom.get_altloc()
            bfac = float(atom.get_bfactor() if atom.get_bfactor() is not None else 0.0)
            occ  = float(atom.get_occupancy() if atom.get_occupancy() is not None else 1.0)
            element = getattr(atom, "element", None)
            atom_tuples.append((name, fullname, altloc, bfac, occ, element))
            coords.append(atom.get_coord().astype(float))
        if atom_tuples:
            residues_template.append((res_id, resname, segid, atom_tuples))
    return residues_template, [np.array(c, dtype=float) for c in coords]

def build_chain_from_template(chain_id, residues_template, coords_iter, serial_start=1):
    ch = Chain.Chain(chain_id)
    serial = serial_start
    idx = 0
    for (res_id, resname, segid, atom_tpls) in residues_template:
        res = Residue.Residue(res_id, resname, segid)
        for (name, fullname, altloc, bfac, occ, element) in atom_tpls:
            coord = coords_iter[idx]; idx += 1
            atm = Atom.Atom(name, coord, bfac, occ, altloc, fullname, serial, element=element)
            serial += 1
            res.add(atm)
        ch.add(res)
    return ch, serial

def centroid(coords):
    X = np.vstack(coords)
    return X.mean(axis=0)

def project_vec_onto_axis(v, u):
    return np.dot(v, u) * u

def orthonormal_basis_from_normal(n):
    """Given unit normal n, build two orthonormal in-plane vectors (a, b)."""
    # Pick a vector not parallel to n
    t = np.array([1.0, 0.0, 0.0])
    if abs(np.dot(t, n)) > 0.9:
        t = np.array([0.0, 1.0, 0.0])
    a = t - np.dot(t, n) * n
    a /= (np.linalg.norm(a) + 1e-12)
    b = np.cross(n, a)
    b /= (np.linalg.norm(b) + 1e-12)
    return a, b

def plane_from_AB(coords_A, coords_B):
    """Return (plane_point m, unit normal n) using PCA on combined A∪B."""
    X = np.vstack([np.vstack(coords_A), np.vstack(coords_B)])
    m = X.mean(axis=0)
    Xc = X - m
    _, _, vt = np.linalg.svd(Xc, full_matrices=False)
    # vt rows are principal axes; third row is normal to best-fit plane
    n = vt[2]
    n = n / (np.linalg.norm(n) + 1e-12)
    return m, n

def solve_center_for_inplane_shift(R, plane_point, n, Acent, d_plane):
    """
    Solve for rotation center c (lying in the plane) so that:
       (R - I) * (Acent - c) = d_plane
    with c constrained to the plane: n·(c - plane_point) = 0
    Parameterize c = plane_point + x*a + y*b, where (a,b) span the plane.
    Solve for x,y in 2D.
    """
    a, b = orthonormal_basis_from_normal(n)
    # Build 3x2 matrix for (R - I) @ (-(x*a + y*b)) = d_plane - (R - I) @ (Acent - plane_point)
    M = (R - np.eye(3))
    rhs = d_plane - M @ (Acent - plane_point)
    A_mat = -np.column_stack([M @ a, M @ b])  # 3x2
    # Least squares for 3x2
    sol, _, _, _ = np.linalg.lstsq(A_mat, rhs, rcond=None)
    x, y = sol[0], sol[1]
    c = plane_point + x * a + y * b
    return c

def apply_transform(coords, R, t):
    return [(R @ p) + t for p in coords]

def diagnose_and_optionally_snap_preserve_plane(R_est, t_est, coords_A, coords_B):
    """
    Keep the estimated A->B transform unless the rotation angle is near 0° or 180°.
    If snapping, construct a C6 transform that preserves the A/B plane:
      - plane = best-fit plane to A ∪ B
      - axis = plane normal
      - angle = 60°
      - choose center c in the plane to match the in-plane centroid shift
      - ensure translation has no component along the normal
    """
    u_est, theta_est = rot_axis_angle(R_est)
    angle_deg = abs(theta_est * 180.0 / pi)

    if (ANGLE_MIN_DEG_FOR_SNAP <= angle_deg <= ANGLE_MAX_DEG_FOR_SNAP):
        return R_est, t_est, False  # keep linear look

    # Snap to C6 about plane normal
    # Get plane (point m in plane, unit normal n)
    m, n = plane_from_AB(coords_A, coords_B)

    # 60° rotation about axis n
    R_c6 = R_from_axis_angle(n, 2.0 * pi / 6.0)

    # Centroid shift between A and B (estimated transform)
    Acent = centroid(coords_A)
    Bcent_est = centroid(apply_transform(coords_A, R_est, t_est))
    d = Bcent_est - Acent

    # In-plane component only (preserve plane)
    d_plane = d - project_vec_onto_axis(d, n)

    # Solve for center c (in plane) so (R_c6 - I)(Acent - c) = d_plane
    c = solve_center_for_inplane_shift(R_c6, m, n, Acent, d_plane)

    # Translation that rotates about c, plus NO normal drift
    t_c6 = c - (R_c6 @ c)

    # Remove any stray normal component from t_c6 (numerical safety)
    t_c6 = t_c6 - project_vec_onto_axis(t_c6, n)

    return R_c6, t_c6, True

def process_cif(cif_path):
    base = os.path.basename(cif_path)
    stem = os.path.splitext(base)[0]
    out_pdb = os.path.join(OUTPUT_DIR, f"{stem}_propagated.pdb")

    parser = MMCIFParser(QUIET=True)
    try:
        structure = parser.get_structure(stem, cif_path)
    except Exception as e:
        print(f"[SKIP] {base}: parse error: {e}")
        return

    model = get_first_model(structure)
    if model is None:
        print(f"[SKIP] {base}: no model")
        return

    chain_A, whyA = get_chain_resolved(structure, CHAIN_FROM)
    chain_B, whyB = get_chain_resolved(structure, CHAIN_TO)
    if chain_A is None or chain_B is None:
        avail = ", ".join(sorted(c.id for c in model.get_chains()))
        print(f"[SKIP] {base}: cannot resolve A/B (A:{whyA}; B:{whyB}). Present: [{avail}]")
        return
    else:
        print(f"[INFO] {base}: A via {whyA}, B via {whyB}")

    # CA correspondence for A->B transform
    A_ca = ca_map(chain_A); B_ca = ca_map(chain_B)
    common = sorted(set(A_ca.keys()).intersection(B_ca.keys()))
    if len(common) < 3:
        print(f"[SKIP] {base}: insufficient CA pairs ({len(common)})")
        return
    P = np.vstack([A_ca[k] for k in common])  # A
    Q = np.vstack([B_ca[k] for k in common])  # B
    R_est, t_est = kabsch(P, Q)

    # Full-atom template/coords from A
    residues_template, coords_A = extract_chain_template(chain_A)
    if not coords_A:
        print(f"[SKIP] {base}: chain A has no atoms")
        return
    coords_B_est = apply_transform(coords_A, R_est, t_est)

    # Decide whether to snap; if so, preserve A/B plane
    R_use, t_use, snapped = diagnose_and_optionally_snap_preserve_plane(R_est, t_est, coords_A, coords_B_est)
    if snapped:
        print(f"[INFO] {base}: snapped to C6 preserving A/B plane (axis = plane normal, angle = 60°).")
    else:
        print(f"[INFO] {base}: using estimated A->B transform (linear propagation).")

    # Build output structure: A original, then B..F by iterative application
    new_struct = Structure.Structure(f"{stem}_propagated")
    new_model = Model.Model(0)

    serial = 1
    # Chain A
    chainA_new, serial = build_chain_from_template(OUTPUT_CHAIN_IDS[0], residues_template, coords_A, serial_start=serial)
    new_model.add(chainA_new)

    # Iteratively apply chosen transform for B..F
    working = [p.copy() for p in coords_A]
    for cid in OUTPUT_CHAIN_IDS[1:]:
        working = apply_transform(working, R_use, t_use)
        chain_new, serial = build_chain_from_template(cid, residues_template, working, serial_start=serial)
        new_model.add(chain_new)

    new_struct.add(new_model)

    io = PDBIO()
    io.set_structure(new_struct)
    try:
        io.save(out_pdb)
        print(f"[OK]  {base} -> {out_pdb}")
    except Exception as e:
        print(f"[FAIL] {base}: write error: {e}")

def main():
    files = sorted(glob.glob(os.path.join(INPUT_DIR, "*.cif")))
    if not files:
        print(f"No CIF files found in {INPUT_DIR}")
        return
    for fp in files:
        process_cif(fp)

if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
"""
Linear-by-default propagation with conditional C6 snap that preserves the A/B plane.

Behavior:
- Default: use the estimated A->B rigid transform (iterative screw-like propagation).
- If A->B is near-linear (tiny angle) or near-180° flip: snap to a cyclic C6.
  * Define the "largest plane" from A and B by PCA (best-fit plane to A∪B).
  * Use the plane normal as the rotation axis.
  * Rotate 60° about a center lying in that plane.
  * Solve for the center so the A->B centroid shift projected in-plane is reproduced.
  * Remove any normal (out-of-plane) translation to keep the plane unchanged.

Inputs:  /home/pkwon2/aragonite/alphafold3/filtered/*.cif
Outputs: /home/pkwon2/aragonite/alphafold3/propagation/<stem>_propagated.pdb
"""

import os
import glob
import numpy as np
from math import acos, cos, sin, pi
from Bio.PDB import MMCIFParser, PDBIO
from Bio.PDB import Atom, Residue, Chain, Model, Structure
from Bio.PDB.Polypeptide import is_aa

INPUT_DIR = "/home/pkwon2/aragonite/alphafold3_alt_sym_axis/filtered"
OUTPUT_DIR = "/home/pkwon2/aragonite/alphafold3_alt_sym_axis/propagation"

CHAIN_FROM = "A"
CHAIN_TO   = "B"
OUTPUT_CHAIN_IDS = ["A", "B", "C", "D", "E", "F"]

# Snap only when rotation is almost linear/flip
ANGLE_MIN_DEG_FOR_SNAP = 15.0    # snap if angle < 15°
ANGLE_MAX_DEG_FOR_SNAP = 165.0   # or if angle > 165°

os.makedirs(OUTPUT_DIR, exist_ok=True)

def kabsch(P, Q):
    Pc = P.mean(axis=0); Qc = Q.mean(axis=0)
    P0, Q0 = P - Pc, Q - Qc
    H = P0.T @ Q0
    U, S, Vt = np.linalg.svd(H)
    R = Vt.T @ U.T
    if np.linalg.det(R) < 0.0:
        Vt[-1, :] *= -1.0
        R = Vt.T @ U.T
    t = Qc - R @ Pc
    return R, t

def rot_axis_angle(R):
    tr = np.clip((np.trace(R) - 1.0) / 2.0, -1.0, 1.0)
    theta = acos(tr)
    ux = R[2,1] - R[1,2]
    uy = R[0,2] - R[2,0]
    uz = R[1,0] - R[0,1]
    u = np.array([ux, uy, uz], dtype=float)
    n = np.linalg.norm(u)
    if n < 1e-8:
        w, v = np.linalg.eig(R)
        idx = np.argmin(np.abs(np.real(w) - 1.0))
        u = np.real(v[:, idx])
        n = np.linalg.norm(u)
        if n < 1e-8:
            u = np.array([0.0, 0.0, 1.0]); n = 1.0
    return u / n, theta

def R_from_axis_angle(u, theta):
    ux, uy, uz = u
    K = np.array([[0, -uz,  uy],
                  [uz,  0, -ux],
                  [-uy, ux,  0]], dtype=float)
    I = np.eye(3)
    return I + sin(theta) * K + (1 - cos(theta)) * (K @ K)

def get_first_model(structure):
    return next(structure.get_models(), None)

def get_chain_resolved(structure, target_id):
    model = get_first_model(structure)
    if model is None:
        return None, "no models found"
    chains = list(model.get_chains())
    if not chains:
        return None, "no chains in model"

    by_exact = {c.id: c for c in chains}
    if target_id in by_exact:
        return by_exact[target_id], f"exact '{target_id}'"

    target_norm = target_id.strip().upper()
    by_norm = {c.id.strip().upper(): c for c in chains}
    if target_norm in by_norm:
        return by_norm[target_norm], f"normalized '{target_norm}'"

    first_map = {}
    for c in chains:
        key = (c.id or "").strip().upper()[:1]
        if key and key not in first_map:
            first_map[key] = c
    if target_norm[:1] in first_map:
        c = first_map[target_norm[:1]]
        return c, f"first-char '{target_norm[:1]}' -> '{c.id}'"

    avail = ", ".join(sorted(c.id for c in chains))
    return None, f"could not resolve '{target_id}'. Available: {avail}"

def ca_map(chain):
    m = {}
    for res in chain:
        if not is_aa(res, standard=True):
            continue
        if "CA" in res:
            resseq = res.id[1]; icode = res.id[2]
            m[(resseq, icode)] = res["CA"].get_coord()
    return m

def extract_chain_template(chain):
    residues_template, coords = [], []
    for res in chain:
        if res.id[0] not in (" ", "A"):
            continue
        res_id = res.id
        resname = res.resname
        segid = getattr(res, "segid", "    ")
        atom_tuples = []
        for atom in res.get_atoms():
            if atom.get_parent().id[0] not in (" ", "A"):
                continue
            name = atom.get_name()
            fullname = atom.get_fullname()
            altloc = atom.get_altloc()
            bfac = float(atom.get_bfactor() if atom.get_bfactor() is not None else 0.0)
            occ  = float(atom.get_occupancy() if atom.get_occupancy() is not None else 1.0)
            element = getattr(atom, "element", None)
            atom_tuples.append((name, fullname, altloc, bfac, occ, element))
            coords.append(atom.get_coord().astype(float))
        if atom_tuples:
            residues_template.append((res_id, resname, segid, atom_tuples))
    return residues_template, [np.array(c, dtype=float) for c in coords]

def build_chain_from_template(chain_id, residues_template, coords_iter, serial_start=1):
    ch = Chain.Chain(chain_id)
    serial = serial_start
    idx = 0
    for (res_id, resname, segid, atom_tpls) in residues_template:
        res = Residue.Residue(res_id, resname, segid)
        for (name, fullname, altloc, bfac, occ, element) in atom_tpls:
            coord = coords_iter[idx]; idx += 1
            atm = Atom.Atom(name, coord, bfac, occ, altloc, fullname, serial, element=element)
            serial += 1
            res.add(atm)
        ch.add(res)
    return ch, serial

def centroid(coords):
    X = np.vstack(coords)
    return X.mean(axis=0)

def project_vec_onto_axis(v, u):
    return np.dot(v, u) * u

def orthonormal_basis_from_normal(n):
    """Given unit normal n, build two orthonormal in-plane vectors (a, b)."""
    # Pick a vector not parallel to n
    t = np.array([1.0, 0.0, 0.0])
    if abs(np.dot(t, n)) > 0.9:
        t = np.array([0.0, 1.0, 0.0])
    a = t - np.dot(t, n) * n
    a /= (np.linalg.norm(a) + 1e-12)
    b = np.cross(n, a)
    b /= (np.linalg.norm(b) + 1e-12)
    return a, b

def plane_from_AB(coords_A, coords_B):
    """Return (plane_point m, unit normal n) using PCA on combined A∪B."""
    X = np.vstack([np.vstack(coords_A), np.vstack(coords_B)])
    m = X.mean(axis=0)
    Xc = X - m
    _, _, vt = np.linalg.svd(Xc, full_matrices=False)
    # vt rows are principal axes; third row is normal to best-fit plane
    n = vt[2]
    n = n / (np.linalg.norm(n) + 1e-12)
    return m, n

def solve_center_for_inplane_shift(R, plane_point, n, Acent, d_plane):
    """
    Solve for rotation center c (lying in the plane) so that:
       (R - I) * (Acent - c) = d_plane
    with c constrained to the plane: n·(c - plane_point) = 0
    Parameterize c = plane_point + x*a + y*b, where (a,b) span the plane.
    Solve for x,y in 2D.
    """
    a, b = orthonormal_basis_from_normal(n)
    # Build 3x2 matrix for (R - I) @ (-(x*a + y*b)) = d_plane - (R - I) @ (Acent - plane_point)
    M = (R - np.eye(3))
    rhs = d_plane - M @ (Acent - plane_point)
    A_mat = -np.column_stack([M @ a, M @ b])  # 3x2
    # Least squares for 3x2
    sol, _, _, _ = np.linalg.lstsq(A_mat, rhs, rcond=None)
    x, y = sol[0], sol[1]
    c = plane_point + x * a + y * b
    return c

def apply_transform(coords, R, t):
    return [(R @ p) + t for p in coords]

def diagnose_and_optionally_snap_preserve_plane(R_est, t_est, coords_A, coords_B):
    """
    Keep the estimated A->B transform unless the rotation angle is near 0° or 180°.
    If snapping, construct a C6 transform that preserves the A/B plane:
      - plane = best-fit plane to A ∪ B
      - axis = plane normal
      - angle = 60°
      - choose center c in the plane to match the in-plane centroid shift
      - ensure translation has no component along the normal
    """
    u_est, theta_est = rot_axis_angle(R_est)
    angle_deg = abs(theta_est * 180.0 / pi)

    if (ANGLE_MIN_DEG_FOR_SNAP <= angle_deg <= ANGLE_MAX_DEG_FOR_SNAP):
        return R_est, t_est, False  # keep linear look

    # Snap to C6 about plane normal
    # Get plane (point m in plane, unit normal n)
    m, n = plane_from_AB(coords_A, coords_B)

    # 60° rotation about axis n
    R_c6 = R_from_axis_angle(n, 2.0 * pi / 6.0)

    # Centroid shift between A and B (estimated transform)
    Acent = centroid(coords_A)
    Bcent_est = centroid(apply_transform(coords_A, R_est, t_est))
    d = Bcent_est - Acent

    # In-plane component only (preserve plane)
    d_plane = d - project_vec_onto_axis(d, n)

    # Solve for center c (in plane) so (R_c6 - I)(Acent - c) = d_plane
    c = solve_center_for_inplane_shift(R_c6, m, n, Acent, d_plane)

    # Translation that rotates about c, plus NO normal drift
    t_c6 = c - (R_c6 @ c)

    # Remove any stray normal component from t_c6 (numerical safety)
    t_c6 = t_c6 - project_vec_onto_axis(t_c6, n)

    return R_c6, t_c6, True

def process_cif(cif_path):
    base = os.path.basename(cif_path)
    stem = os.path.splitext(base)[0]
    out_pdb = os.path.join(OUTPUT_DIR, f"{stem}_propagated.pdb")

    parser = MMCIFParser(QUIET=True)
    try:
        structure = parser.get_structure(stem, cif_path)
    except Exception as e:
        print(f"[SKIP] {base}: parse error: {e}")
        return

    model = get_first_model(structure)
    if model is None:
        print(f"[SKIP] {base}: no model")
        return

    chain_A, whyA = get_chain_resolved(structure, CHAIN_FROM)
    chain_B, whyB = get_chain_resolved(structure, CHAIN_TO)
    if chain_A is None or chain_B is None:
        avail = ", ".join(sorted(c.id for c in model.get_chains()))
        print(f"[SKIP] {base}: cannot resolve A/B (A:{whyA}; B:{whyB}). Present: [{avail}]")
        return
    else:
        print(f"[INFO] {base}: A via {whyA}, B via {whyB}")

    # CA correspondence for A->B transform
    A_ca = ca_map(chain_A); B_ca = ca_map(chain_B)
    common = sorted(set(A_ca.keys()).intersection(B_ca.keys()))
    if len(common) < 3:
        print(f"[SKIP] {base}: insufficient CA pairs ({len(common)})")
        return
    P = np.vstack([A_ca[k] for k in common])  # A
    Q = np.vstack([B_ca[k] for k in common])  # B
    R_est, t_est = kabsch(P, Q)

    # Full-atom template/coords from A
    residues_template, coords_A = extract_chain_template(chain_A)
    if not coords_A:
        print(f"[SKIP] {base}: chain A has no atoms")
        return
    coords_B_est = apply_transform(coords_A, R_est, t_est)

    # Decide whether to snap; if so, preserve A/B plane
    R_use, t_use, snapped = diagnose_and_optionally_snap_preserve_plane(R_est, t_est, coords_A, coords_B_est)
    if snapped:
        print(f"[INFO] {base}: snapped to C6 preserving A/B plane (axis = plane normal, angle = 60°).")
    else:
        print(f"[INFO] {base}: using estimated A->B transform (linear propagation).")

    # Build output structure: A original, then B..F by iterative application
    new_struct = Structure.Structure(f"{stem}_propagated")
    new_model = Model.Model(0)

    serial = 1
    # Chain A
    chainA_new, serial = build_chain_from_template(OUTPUT_CHAIN_IDS[0], residues_template, coords_A, serial_start=serial)
    new_model.add(chainA_new)

    # Iteratively apply chosen transform for B..F
    working = [p.copy() for p in coords_A]
    for cid in OUTPUT_CHAIN_IDS[1:]:
        working = apply_transform(working, R_use, t_use)
        chain_new, serial = build_chain_from_template(cid, residues_template, working, serial_start=serial)
        new_model.add(chain_new)

    new_struct.add(new_model)

    io = PDBIO()
    io.set_structure(new_struct)
    try:
        io.save(out_pdb)
        print(f"[OK]  {base} -> {out_pdb}")
    except Exception as e:
        print(f"[FAIL] {base}: write error: {e}")

def main():
    files = sorted(glob.glob(os.path.join(INPUT_DIR, "*.cif")))
    if not files:
        print(f"No CIF files found in {INPUT_DIR}")
        return
    for fp in files:
        process_cif(fp)

if __name__ == "__main__":
    main()
```

```python
import os
import glob
from Bio.PDB import MMCIFParser, PPBuilder
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO
from Bio.Seq import Seq

# Directory containing CIF files
glob_dir = '/home/pkwon2/aragonite/alphafold3_alt_sym_axis/filtered'
# Output FASTA file
fasta_file = os.path.join(glob_dir, 'chainA_first_third.fasta')
# Optional sequence to prepend before each chain A sequence
head_seq = ""

# Initialize the CIF parser and peptide builder
parser = MMCIFParser(QUIET=True)
ppb = PPBuilder()
records = []

# Loop over all .cif files in the directory
for cif_path in glob.glob(os.path.join(glob_dir, '*.cif')):
    pdb_id = os.path.basename(cif_path).split('.')[0]
    
    try:
        structure = parser.get_structure(pdb_id, cif_path)
    except Exception as e:
        print(f"Error parsing {cif_path}: {e}")
        continue

    model = structure[0]
    if 'A' not in model:
        print(f"Chain A not found in {cif_path}")
        continue
    chain_a = model['A']

    # Build polypeptides (handles discontinuous segments)
    peptides = ppb.build_peptides(chain_a)
    seq_str = "".join(str(pp.get_sequence()) for pp in peptides)

    if not seq_str:
        print(f"No sequence extracted from chain A in {cif_path}")
        continue

    # Keep only the first third of the sequence
    first_third_len = max(1, len(seq_str) // 3)
    trimmed_seq = seq_str[:first_third_len]

    # Prepend head sequence and create FASTA record
    full_seq = head_seq + trimmed_seq
    record = SeqRecord(Seq(full_seq), id=f"{pdb_id}_A_first_third", description="")
    records.append(record)
    print(f"Extracted first third of chain A from {pdb_id} ({first_third_len} residues).")

# Write all records to the FASTA file
if records:
    SeqIO.write(records, fasta_file, 'fasta')
    print(f"\nWrote {len(records)} record(s) to {fasta_file}")
else:
    print("No chain A sequences were extracted.")
```

```python
import os
from Bio import SeqIO

def create_directory_structure(base_directory):
    """
    Creates a subdirectory named 'fasta_files' within the base_directory.
    Returns the full path to the created subdirectory.
    """
    output_directory = os.path.join(base_directory, "fasta_files")
    if not os.path.exists(output_directory):
        os.makedirs(output_directory)
    return output_directory

def read_fasta(file_path):
    """
    Reads sequences from a FASTA file and returns a list of dictionaries,
    each containing the sequence name and its sequence string.
    """
    sequences = []
    for record in SeqIO.parse(file_path, "fasta"):
        sequences.append({'name': record.id, 'sequence': str(record.seq)})
    return sequences

def write_fasta_files(sequences, output_directory):
    """
    For each sequence, writes a separate FASTA file with:
    [sequence] + " " + [sequence] + " " + [sequence]
    """
    for seq in sequences:
        # Join three copies separated by spaces
        fasta_string = " ".join([seq['sequence']] * 3)
        fasta_content = f">{seq['name']}\n{fasta_string}\n"
        
        fasta_file_path = os.path.join(output_directory, f"{seq['name']}.fasta")
        with open(fasta_file_path, 'w') as f:
            f.write(fasta_content)
        print(f"Created FASTA file: {fasta_file_path}")

if __name__ == "__main__":
    # Define base and output paths
    base_directory = "alphafold3"
    output_fasta_directory = create_directory_structure(base_directory)
    
    # Define the input FASTA file path
    input_fasta_path = '/home/pkwon2/aragonite/alphafold3_alt_sym_axis/filtered/chainA_first_third.fasta'
    
    # Read sequences from the FASTA file
    sequences = read_fasta(input_fasta_path)
    
    # Write modified FASTA files
    write_fasta_files(sequences, output_fasta_directory)
```
