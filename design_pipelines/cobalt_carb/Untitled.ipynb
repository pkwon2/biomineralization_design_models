```python
#!/usr/bin/env python3
import csv
import textwrap

# --- Input sequences (C3 first, then C2) ---
C3 = {
    "c3_1_1":  """AEKEKATLTAVLTDVLRTNLATGSVTVRAATTLLRLTIAFKATEGVDAALATVREALALLAARPTKTARDTDLTADAAKEGADLAAEHGRAQDAITLAREAATLLAKANAV""",
    "c3_5_1":  """RAEAFATLREVLRDALLTNVATGSVTVETARTLLRLTIAFDATVGDDEALATVRDVLATLAARPTLTARDTDLTADAAKEGADLAAEHGRAQDAITLAREAATLLAKANAV""",
    "c3_10_1": """REEEFQRLLDVVRDTLLANLASGSVTARAARTLLLLGIALAARVGADAALALVREALATLAARPTKTALDTDLTADAAKEGADLAAEHGRAQDAITLAREAATLLAKANAV""",
}

C2 = {
    "c2_11_1": """DGLLDLADDLADHGLIEGAVAVFRNAIDAAAAAGLADHTADAVLEAVDLAAEHGDAEGLVAVAREGLRALDALGADDPDVRAVCLAAAALGEARLALERGLDRREVLRRIRALLDELEALAGVSDALARRVAALLLALTRLLVESELA""",
    "c2_12_1": """DGLLDLADDLADHGLIEGAVAVFRNAIDAAAAAGLADHTADAVLEAVDLAAEHGDAEGLLAVAREGLRALDALGADDADTRAELLAAAALATARLLVERGLDLREALRRIRALLDELAALAGISEARARRVAALLLDTVRLLVELLLA""",
}

FASTA_OUT = "c3x_c2_combinations.fasta"
CSV_OUT   = "c3x_c2_combinations.csv"
LINE_WRAP = 80  # characters per FASTA line

def clean(seq: str) -> str:
    return "".join(seq.split()).upper()

def write_fasta(records, path, wrap=80):
    with open(path, "w") as f:
        for header, seq in records:
            f.write(f">{header}\n")
            if wrap and wrap > 0:
                for chunk in textwrap.wrap(seq, wrap):
                    f.write(chunk + "\n")
            else:
                f.write(seq + "\n")

def main():
    # Clean sequences (remove spaces/newlines)
    c3_clean = {k: clean(v) for k, v in C3.items()}
    c2_clean = {k: clean(v) for k, v in C2.items()}

    fasta_records = []
    csv_rows = []

    for c3_name, c3_seq in c3_clean.items():
        for c2_name, c2_seq in c2_clean.items():
            combo_name = f"{c3_name}__{c2_name}"
            combo_seq  = c3_seq + c2_seq
            fasta_records.append((combo_name, combo_seq))
            csv_rows.append({
                "combo_name": combo_name,
                "c3_name": c3_name,
                "c2_name": c2_name,
                "c3_len": len(c3_seq),
                "c2_len": len(c2_seq),
                "combo_len": len(combo_seq),
            })

    # Write FASTA
    write_fasta(fasta_records, FASTA_OUT, wrap=LINE_WRAP)

    # Write CSV summary
    with open(CSV_OUT, "w", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=["combo_name","c3_name","c2_name","c3_len","c2_len","combo_len"])
        writer.writeheader()
        writer.writerows(csv_rows)

    print(f"Wrote {len(fasta_records)} combinations to {FASTA_OUT}")
    print(f"Wrote summary to {CSV_OUT}")

if __name__ == "__main__":
    main()
```
